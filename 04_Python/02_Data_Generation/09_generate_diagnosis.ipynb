```python
# ==============================================================================
# IMPORT REQUIRED LIBRARIES
# ==============================================================================

import os
import random
import numpy as np
import pandas as pd

from faker import Faker
from datetime import datetime


# ==============================================================================
# INITIALIZE RANDOM GENERATORS
# ==============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

fake = Faker("en_IN")
Faker.seed(SEED)

print("Libraries imported successfully.")
print(f"Random seed: {SEED}")
```

```python
# ==============================================================================
# DIAGNOSIS GENERATION CONFIGURATION
# ==============================================================================

TOTAL_DIAGNOSES = 300

OUTPUT_DIRECTORY = os.path.join(
    "03_Datasets",
    "CSV_Files"
)

CSV_FILE_PATH = os.path.join(
    OUTPUT_DIRECTORY,
    "dim_diagnosis.csv"
)

TODAY = datetime.now()


print("=" * 80)
print("DIAGNOSIS GENERATION CONFIGURATION")
print("=" * 80)

print(
    f"Total diagnoses : {TOTAL_DIAGNOSES}"
)

print(
    f"Output path     : {CSV_FILE_PATH}"
)
```

```python
# ==============================================================================
# DIAGNOSIS MASTER DATA
# ==============================================================================

DIAGNOSIS_CATEGORIES = [
    "Infectious Disease",
    "Cardiovascular",
    "Respiratory",
    "Gastrointestinal",
    "Neurological",
    "Endocrine",
    "Musculoskeletal",
    "Dermatological",
    "Genitourinary",
    "Oncology",
    "Hematological",
    "Renal",
    "Hepatic",
    "Ophthalmic",
    "Psychiatric",
    "Pediatric",
    "Gynecological",
    "Trauma",
    "Autoimmune",
    "Metabolic"
]


DIAGNOSIS_TYPES = [
    "Primary",
    "Secondary",
    "Provisional",
    "Chronic",
    "Acute",
    "Comorbidity"
]


SEVERITY_LEVELS = [
    "Mild",
    "Moderate",
    "Severe",
    "Critical"
]


BODY_SYSTEMS = [
    "Cardiovascular System",
    "Respiratory System",
    "Digestive System",
    "Nervous System",
    "Endocrine System",
    "Musculoskeletal System",
    "Integumentary System",
    "Urinary System",
    "Reproductive System",
    "Immune System",
    "Hematological System",
    "Visual System"
]


DEPARTMENTS = [
    "General Medicine",
    "Cardiology",
    "Pulmonology",
    "Neurology",
    "Gastroenterology",
    "Endocrinology",
    "Orthopedics",
    "Dermatology",
    "Nephrology",
    "Urology",
    "Oncology",
    "Pediatrics",
    "Gynecology",
    "Emergency Medicine",
    "Psychiatry"
]


DIAGNOSIS_STATUSES = [
    "Active",
    "Inactive",
    "Deprecated"
]


print("Diagnosis master data loaded successfully.")
```

```python
# ==============================================================================
# DIAGNOSIS NAME MASTER DATA
# ==============================================================================

DIAGNOSIS_NAMES = [
    "Essential Hypertension",
    "Type 2 Diabetes Mellitus",
    "Type 1 Diabetes Mellitus",
    "Coronary Artery Disease",
    "Acute Myocardial Infarction",
    "Heart Failure",
    "Atrial Fibrillation",
    "Hyperlipidemia",
    "Angina Pectoris",
    "Cardiomyopathy",
    "Asthma",
    "Chronic Obstructive Pulmonary Disease",
    "Pneumonia",
    "Acute Bronchitis",
    "Pulmonary Tuberculosis",
    "Allergic Rhinitis",
    "Respiratory Tract Infection",
    "Gastroesophageal Reflux Disease",
    "Peptic Ulcer Disease",
    "Acute Gastritis",
    "Irritable Bowel Syndrome",
    "Acute Appendicitis",
    "Cholelithiasis",
    "Acute Pancreatitis",
    "Fatty Liver Disease",
    "Hepatitis B",
    "Hepatitis C",
    "Cirrhosis of Liver",
    "Chronic Kidney Disease",
    "Acute Kidney Injury",
    "Urinary Tract Infection",
    "Kidney Stone",
    "Nephrotic Syndrome",
    "Migraine",
    "Epilepsy",
    "Ischemic Stroke",
    "Hemorrhagic Stroke",
    "Parkinson's Disease",
    "Alzheimer's Disease",
    "Peripheral Neuropathy",
    "Osteoarthritis",
    "Rheumatoid Arthritis",
    "Osteoporosis",
    "Fracture",
    "Low Back Pain",
    "Gout",
    "Psoriasis",
    "Eczema",
    "Dermatitis",
    "Acne Vulgaris",
    "Anemia",
    "Iron Deficiency Anemia",
    "Thrombocytopenia",
    "Leukemia",
    "Lymphoma",
    "Breast Cancer",
    "Lung Cancer",
    "Colorectal Cancer",
    "Prostate Cancer",
    "Thyroid Cancer",
    "Thyroid Disorder",
    "Hypothyroidism",
    "Hyperthyroidism",
    "Obesity",
    "Vitamin D Deficiency",
    "Vitamin B12 Deficiency",
    "Polycystic Ovary Syndrome",
    "Endometriosis",
    "Pelvic Inflammatory Disease",
    "Depression",
    "Generalized Anxiety Disorder",
    "Bipolar Disorder",
    "Schizophrenia",
    "Insomnia",
    "Viral Fever",
    "Dengue Fever",
    "Malaria",
    "Typhoid Fever",
    "Influenza",
    "COVID-19",
    "Hepatitis A",
    "HIV Infection",
    "Sepsis",
    "Meningitis",
    "Osteomyelitis",
    "Conjunctivitis",
    "Glaucoma",
    "Cataract",
    "Diabetic Retinopathy"
]


print(
    f"Available diagnosis names: {len(DIAGNOSIS_NAMES)}"
)
```

```python
# ==============================================================================
# ICD CODE GENERATOR
# ==============================================================================

def generate_icd_code(diagnosis_id):

    category_code = random.choice(
        [
            "E",
            "I",
            "J",
            "K",
            "G",
            "M",
            "N",
            "C",
            "D",
            "L",
            "F"
        ]
    )

    numeric_code = random.randint(
        00,
        99
    )

    decimal_code = random.randint(
        0,
        9
    )

    return (
        f"{category_code}"
        f"{numeric_code:02d}."
        f"{decimal_code}"
    )


print(
    generate_icd_code(1)
)
```

```python
# ==============================================================================
# DIAGNOSIS CODE GENERATOR
# ==============================================================================

def generate_diagnosis_code(diagnosis_id):

    return f"DX{diagnosis_id:05d}"


print(
    generate_diagnosis_code(1)
)

print(
    generate_diagnosis_code(
        TOTAL_DIAGNOSES
    )
)
```

```python
# ==============================================================================
# GENERATE DIAGNOSIS RECORDS
# ==============================================================================

diagnosis_records = []


for diagnosis_id in range(
    1,
    TOTAL_DIAGNOSES + 1
):

    # --------------------------------------------------------------------------
    # IDENTIFIERS
    # --------------------------------------------------------------------------

    diagnosis_code = generate_diagnosis_code(
        diagnosis_id
    )

    icd_code = generate_icd_code(
        diagnosis_id
    )


    # --------------------------------------------------------------------------
    # DIAGNOSIS NAME
    # --------------------------------------------------------------------------

    base_diagnosis_name = random.choice(
        DIAGNOSIS_NAMES
    )

    diagnosis_name = (
        base_diagnosis_name
        + " - "
        + random.choice(
            [
                "Condition",
                "Disorder",
                "Disease",
                "Syndrome",
                "Case"
            ]
        )
    )


    # --------------------------------------------------------------------------
    # CATEGORY
    # --------------------------------------------------------------------------

    diagnosis_category = random.choice(
        DIAGNOSIS_CATEGORIES
    )


    # --------------------------------------------------------------------------
    # DIAGNOSIS TYPE
    # --------------------------------------------------------------------------

    diagnosis_type = random.choice(
        DIAGNOSIS_TYPES
    )


    # --------------------------------------------------------------------------
    # SEVERITY
    # --------------------------------------------------------------------------

    severity = random.choices(
        SEVERITY_LEVELS,
        weights=[
            40,
            40,
            15,
            5
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # CHRONIC CONDITION
    # --------------------------------------------------------------------------

    chronic_condition = random.choices(
        [
            True,
            False
        ],
        weights=[
            35,
            65
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # COMMUNICABLE CONDITION
    # --------------------------------------------------------------------------

    communicable = random.choices(
        [
            True,
            False
        ],
        weights=[
            15,
            85
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # BODY SYSTEM
    # --------------------------------------------------------------------------

    body_system = random.choice(
        BODY_SYSTEMS
    )


    # --------------------------------------------------------------------------
    # DEPARTMENT
    # --------------------------------------------------------------------------

    department = random.choice(
        DEPARTMENTS
    )


    # --------------------------------------------------------------------------
    # DESCRIPTION
    # --------------------------------------------------------------------------

    description = (
        f"Clinical diagnosis associated with "
        f"{diagnosis_category.lower()} conditions "
        f"and requiring appropriate medical evaluation."
    )


    # --------------------------------------------------------------------------
    # STATUS
    # --------------------------------------------------------------------------

    diagnosis_status = random.choices(
        DIAGNOSIS_STATUSES,
        weights=[
            95,
            3,
            2
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # AUDIT INFORMATION
    # --------------------------------------------------------------------------

    created_date = fake.date_time_between(
        start_date="-5y",
        end_date=TODAY
    )

    last_updated = fake.date_time_between(
        start_date=created_date,
        end_date=TODAY
    )


    # --------------------------------------------------------------------------
    # CREATE RECORD
    # --------------------------------------------------------------------------

    diagnosis_records.append(
        {
            "diagnosis_id": diagnosis_id,
            "diagnosis_code": diagnosis_code,
            "diagnosis_name": diagnosis_name,
            "diagnosis_category": diagnosis_category,
            "diagnosis_type": diagnosis_type,
            "icd_code": icd_code,
            "severity": severity,
            "chronic_condition": chronic_condition,
            "communicable": communicable,
            "body_system": body_system,
            "department": department,
            "description": description,
            "diagnosis_status": diagnosis_status,
            "created_date": created_date,
            "last_updated": last_updated
        }
    )


print(
    f"Generated {len(diagnosis_records):,} "
    f"diagnosis records."
)
```

```python
# ==============================================================================
# CREATE DIAGNOSIS DATAFRAME
# ==============================================================================

diagnosis_df = pd.DataFrame(
    diagnosis_records
)


date_columns = [
    "created_date",
    "last_updated"
]


for column in date_columns:

    diagnosis_df[column] = pd.to_datetime(
        diagnosis_df[column]
    )


print("=" * 80)
print("DIAGNOSIS DATAFRAME CREATED")
print("=" * 80)


print(
    f"Rows    : {diagnosis_df.shape[0]:,}"
)


print(
    f"Columns : {diagnosis_df.shape[1]}"
)


diagnosis_df.head()
```

```python
# ==============================================================================
# DIAGNOSIS DATA VALIDATION
# ==============================================================================

print("=" * 80)
print("DIAGNOSIS DATA VALIDATION")
print("=" * 80)


# ------------------------------------------------------------------------------
# 1. Row count
# ------------------------------------------------------------------------------

assert (
    len(diagnosis_df)
    == TOTAL_DIAGNOSES
)

print(
    "✓ Row count validation passed."
)


# ------------------------------------------------------------------------------
# 2. Diagnosis ID uniqueness
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "diagnosis_id"
].is_unique

print(
    "✓ Diagnosis ID uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 3. Diagnosis code uniqueness
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "diagnosis_code"
].is_unique

print(
    "✓ Diagnosis code uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 4. Diagnosis name uniqueness
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "diagnosis_name"
].is_unique

print(
    "✓ Diagnosis name uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 5. ICD code uniqueness
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "icd_code"
].is_unique

print(
    "✓ ICD code uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 6. Required fields
# ------------------------------------------------------------------------------

required_columns = [
    "diagnosis_id",
    "diagnosis_code",
    "diagnosis_name",
    "diagnosis_category",
    "diagnosis_type",
    "icd_code",
    "severity",
    "chronic_condition",
    "communicable",
    "body_system",
    "department",
    "description",
    "diagnosis_status"
]


for column in required_columns:

    assert diagnosis_df[
        column
    ].notna().all(), (
        f"Null values found in {column}"
    )


print(
    "✓ Required field validation passed."
)


# ------------------------------------------------------------------------------
# 7. Category validation
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "diagnosis_category"
].isin(
    DIAGNOSIS_CATEGORIES
).all()

print(
    "✓ Diagnosis category validation passed."
)


# ------------------------------------------------------------------------------
# 8. Diagnosis type validation
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "diagnosis_type"
].isin(
    DIAGNOSIS_TYPES
).all()

print(
    "✓ Diagnosis type validation passed."
)


# ------------------------------------------------------------------------------
# 9. Severity validation
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "severity"
].isin(
    SEVERITY_LEVELS
).all()

print(
    "✓ Severity validation passed."
)


# ------------------------------------------------------------------------------
# 10. Body system validation
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "body_system"
].isin(
    BODY_SYSTEMS
).all()

print(
    "✓ Body system validation passed."
)


# ------------------------------------------------------------------------------
# 11. Department validation
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "department"
].isin(
    DEPARTMENTS
).all()

print(
    "✓ Department validation passed."
)


# ------------------------------------------------------------------------------
# 12. Boolean validation
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "chronic_condition"
].isin(
    [
        True,
        False
    ]
).all()


assert diagnosis_df[
    "communicable"
].isin(
    [
        True,
        False
    ]
).all()

print(
    "✓ Boolean field validation passed."
)


# ------------------------------------------------------------------------------
# 13. Status validation
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "diagnosis_status"
].isin(
    DIAGNOSIS_STATUSES
).all()

print(
    "✓ Diagnosis status validation passed."
)


# ------------------------------------------------------------------------------
# 14. Diagnosis code format
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "diagnosis_code"
].str.match(
    r"^DX\d{5}$"
).all()

print(
    "✓ Diagnosis code format validation passed."
)


# ------------------------------------------------------------------------------
# 15. ICD code format
# ------------------------------------------------------------------------------

assert diagnosis_df[
    "icd_code"
].str.match(
    r"^[A-Z]\d{2}\.\d$"
).all()

print(
    "✓ ICD code format validation passed."
)


print()

print(
    "All diagnosis data validations passed successfully. ✓"
)
```

```python
# ==============================================================================
# EXPORT DIAGNOSIS DATA TO CSV
# ==============================================================================

os.makedirs(
    OUTPUT_DIRECTORY,
    exist_ok=True
)


def export_to_csv(diagnosis_df):
    """
    Export diagnosis dimension data into CSV.
    """

    diagnosis_df.to_csv(
        CSV_FILE_PATH,
        index=False
    )


    print("=" * 80)
    print("DIAGNOSIS DATA EXPORT")
    print("=" * 80)


    print(
        "CSV file created successfully:"
    )


    print(
        CSV_FILE_PATH
    )


    return CSV_FILE_PATH


csv_file = export_to_csv(
    diagnosis_df
)
```
