```python
# ==============================================================================
# IMPORT REQUIRED LIBRARIES
# ==============================================================================

import os
import random
import numpy as np
import pandas as pd

from datetime import datetime, date


# ==============================================================================
# INITIALIZE RANDOM GENERATORS
# ==============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)


print("Libraries imported successfully.")
print(f"Random seed: {SEED}")
```

```python
# ==============================================================================
# DATE DIMENSION CONFIGURATION
# ==============================================================================

START_DATE = pd.Timestamp(
    "2020-01-01"
)

END_DATE = pd.Timestamp(
    "2030-12-31"
)


OUTPUT_DIRECTORY = os.path.join(
    "03_Datasets",
    "CSV_Files"
)


CSV_FILE_PATH = os.path.join(
    OUTPUT_DIRECTORY,
    "dim_date.csv"
)


print("=" * 80)
print("DATE DIMENSION CONFIGURATION")
print("=" * 80)


print(
    f"Start date : {START_DATE.date()}"
)


print(
    f"End date   : {END_DATE.date()}"
)


print(
    f"Output path: {CSV_FILE_PATH}"
)
```

```python
# ==============================================================================
# GENERATE DATE RANGE
# ==============================================================================

date_range = pd.date_range(
    start=START_DATE,
    end=END_DATE,
    freq="D"
)


print("=" * 80)
print("DATE RANGE GENERATED")
print("=" * 80)


print(
    f"Total dates: {len(date_range):,}"
)


print(
    f"First date : {date_range.min().date()}"
)


print(
    f"Last date  : {date_range.max().date()}"
)
```

```python
# ==============================================================================
# GENERATE DATE RECORDS
# ==============================================================================

date_records = []


for current_date in date_range:

    # --------------------------------------------------------------------------
    # BASIC DATE INFORMATION
    # --------------------------------------------------------------------------

    date_id = int(
        current_date.strftime("%Y%m%d")
    )

    full_date = current_date.date()

    day = current_date.day

    month = current_date.month

    year = current_date.year


    # --------------------------------------------------------------------------
    # DAY INFORMATION
    # --------------------------------------------------------------------------

    day_of_week = current_date.dayofweek

    day_name = current_date.strftime(
        "%A"
    )

    day_short_name = current_date.strftime(
        "%a"
    )


    # Monday = 1 ... Sunday = 7

    day_of_week_number = (
        day_of_week + 1
    )


    # --------------------------------------------------------------------------
    # WEEK INFORMATION
    # --------------------------------------------------------------------------

    week_of_year = int(
        current_date.isocalendar().week
    )


    # --------------------------------------------------------------------------
    # MONTH INFORMATION
    # --------------------------------------------------------------------------

    month_name = current_date.strftime(
        "%B"
    )

    month_short_name = current_date.strftime(
        "%b"
    )


    # Month number within quarter

    month_of_quarter = (
        ((month - 1) % 3) + 1
    )


    # --------------------------------------------------------------------------
    # QUARTER INFORMATION
    # --------------------------------------------------------------------------

    quarter_number = (
        ((month - 1) // 3) + 1
    )

    quarter_name = (
        f"Q{quarter_number}"
    )


    # --------------------------------------------------------------------------
    # YEAR-MONTH INFORMATION
    # --------------------------------------------------------------------------

    year_month = (
        current_date.strftime("%Y-%m")
    )

    year_month_number = (
        year * 100 + month
    )


    # --------------------------------------------------------------------------
    # WEEKEND / WEEKDAY
    # --------------------------------------------------------------------------

    is_weekend = (
        day_of_week_number >= 6
    )

    is_weekday = (
        not is_weekend
    )


    # --------------------------------------------------------------------------
    # MONTH START / END
    # --------------------------------------------------------------------------

    is_month_start = (
        current_date.is_month_start
    )

    is_month_end = (
        current_date.is_month_end
    )


    # --------------------------------------------------------------------------
    # QUARTER START / END
    # --------------------------------------------------------------------------

    is_quarter_start = (
        current_date.is_quarter_start
    )

    is_quarter_end = (
        current_date.is_quarter_end
    )


    # --------------------------------------------------------------------------
    # YEAR START / END
    # --------------------------------------------------------------------------

    is_year_start = (
        current_date.is_year_start
    )

    is_year_end = (
        current_date.is_year_end
    )


    # --------------------------------------------------------------------------
    # CREATE RECORD
    # --------------------------------------------------------------------------

    date_records.append(
        {
            "date_id": date_id,
            "full_date": full_date,
            "day": day,
            "day_of_week_number": day_of_week_number,
            "day_name": day_name,
            "day_short_name": day_short_name,
            "week_of_year": week_of_year,
            "month": month,
            "month_name": month_name,
            "month_short_name": month_short_name,
            "month_of_quarter": month_of_quarter,
            "quarter_number": quarter_number,
            "quarter_name": quarter_name,
            "year": year,
            "year_month": year_month,
            "year_month_number": year_month_number,
            "is_weekday": is_weekday,
            "is_weekend": is_weekend,
            "is_month_start": is_month_start,
            "is_month_end": is_month_end,
            "is_quarter_start": is_quarter_start,
            "is_quarter_end": is_quarter_end,
            "is_year_start": is_year_start,
            "is_year_end": is_year_end
        }
    )


print(
    f"Generated {len(date_records):,} "
    f"date records."
)
```

```python
# ==============================================================================
# CREATE DATE DATAFRAME
# ==============================================================================

if len(date_records) == 0:

    raise ValueError(
        "date_records is empty. "
        "Please run the date generation cell first."
    )


dim_date = pd.DataFrame(
    date_records
)


# ------------------------------------------------------------------------------
# CONVERT DATE COLUMN
# ------------------------------------------------------------------------------

dim_date["full_date"] = pd.to_datetime(
    dim_date["full_date"]
)


print("=" * 80)
print("DATE DATAFRAME CREATED")
print("=" * 80)


print(
    f"Rows    : {dim_date.shape[0]:,}"
)


print(
    f"Columns : {dim_date.shape[1]}"
)


dim_date.head()
```

```python
# ==============================================================================
# DATA TYPE CHECK
# ==============================================================================

print("=" * 80)
print("DATE DATA TYPES")
print("=" * 80)


print(
    dim_date.dtypes
)
```

```python
# ==============================================================================
# DATE DIMENSION VALIDATION
# ==============================================================================

print("=" * 80)
print("DATE DIMENSION VALIDATION")
print("=" * 80)


# ------------------------------------------------------------------------------
# 1. Row count
# ------------------------------------------------------------------------------

expected_date_count = (
    (END_DATE - START_DATE).days + 1
)


assert (
    len(dim_date)
    == expected_date_count
)


print(
    "✓ Row count validation passed."
)


# ------------------------------------------------------------------------------
# 2. Date ID uniqueness
# ------------------------------------------------------------------------------

assert dim_date[
    "date_id"
].is_unique


print(
    "✓ Date ID uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 3. Full date uniqueness
# ------------------------------------------------------------------------------

assert dim_date[
    "full_date"
].is_unique


print(
    "✓ Full date uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 4. Date ordering
# ------------------------------------------------------------------------------

assert (
    dim_date[
        "full_date"
    ].min()
    == START_DATE
)


assert (
    dim_date[
        "full_date"
    ].max()
    == END_DATE
)


print(
    "✓ Date range validation passed."
)


# ------------------------------------------------------------------------------
# 5. Date ID format
# ------------------------------------------------------------------------------

assert (
    dim_date[
        "date_id"
    ].astype(str)
    .str.match(
        r"^\d{8}$"
    )
).all()


print(
    "✓ Date ID format validation passed."
)


# ------------------------------------------------------------------------------
# 6. Month validation
# ------------------------------------------------------------------------------

assert dim_date[
    "month"
].between(
    1,
    12
).all()


print(
    "✓ Month validation passed."
)


# ------------------------------------------------------------------------------
# 7. Quarter validation
# ------------------------------------------------------------------------------

assert dim_date[
    "quarter_number"
].between(
    1,
    4
).all()


print(
    "✓ Quarter validation passed."
)


# ------------------------------------------------------------------------------
# 8. Day-of-week validation
# ------------------------------------------------------------------------------

assert dim_date[
    "day_of_week_number"
].between(
    1,
    7
).all()


print(
    "✓ Day-of-week validation passed."
)


# ------------------------------------------------------------------------------
# 9. Boolean validation
# ------------------------------------------------------------------------------

boolean_columns = [
    "is_weekday",
    "is_weekend",
    "is_month_start",
    "is_month_end",
    "is_quarter_start",
    "is_quarter_end",
    "is_year_start",
    "is_year_end"
]


for column in boolean_columns:

    assert dim_date[
        column
    ].isin(
        [
            True,
            False
        ]
    ).all()


print(
    "✓ Boolean field validation passed."
)


# ------------------------------------------------------------------------------
# 10. Null validation
# ------------------------------------------------------------------------------

assert (
    dim_date.isnull().sum().sum()
    == 0
)


print(
    "✓ Null value validation passed."
)


print()

print(
    "All date dimension validations "
    "passed successfully. ✓"
)
```

```python
# ==============================================================================
# CHECK DUPLICATE RECORDS
# ==============================================================================

duplicate_count = (
    dim_date
    .duplicated()
    .sum()
)


print("=" * 80)
print("DUPLICATE CHECK")
print("=" * 80)


print(
    f"Duplicate records: {duplicate_count}"
)


assert duplicate_count == 0


print(
    "✓ No duplicate records found."
)# ==============================================================================
# CHECK DUPLICATE RECORDS
# ==============================================================================

duplicate_count = (
    dim_date
    .duplicated()
    .sum()
)


print("=" * 80)
print("DUPLICATE CHECK")
print("=" * 80)


print(
    f"Duplicate records: {duplicate_count}"
)


assert duplicate_count == 0


print(
    "✓ No duplicate records found."
)
```

```python
# ==============================================================================
# DATA QUALITY SUMMARY
# ==============================================================================

print("=" * 80)
print("DATA QUALITY SUMMARY")
print("=" * 80)


quality_summary = pd.DataFrame(
    {
        "Metric": [
            "Total Dates",
            "Total Columns",
            "Duplicate Rows",
            "Null Values",
            "Unique Date IDs",
            "Unique Full Dates",
            "Total Years",
            "Total Months",
            "Total Quarters",
            "Weekdays",
            "Weekends"
        ],

        "Value": [
            len(dim_date),
            len(dim_date.columns),
            dim_date.duplicated().sum(),
            dim_date.isnull().sum().sum(),
            dim_date["date_id"].nunique(),
            dim_date["full_date"].nunique(),
            dim_date["year"].nunique(),
            dim_date["year_month"].nunique(),
            (
                dim_date[
                    [
                        "year",
                        "quarter_number"
                    ]
                ]
                .drop_duplicates()
                .shape[0]
            ),
            dim_date["is_weekday"].sum(),
            dim_date["is_weekend"].sum()
        ]
    }
)


quality_summary
```

```python
# ==============================================================================
# YEAR DISTRIBUTION
# ==============================================================================

print("=" * 80)
print("YEAR DISTRIBUTION")
print("=" * 80)


year_summary = (
    dim_date
    .groupby("year")
    .agg(
        total_days=(
            "date_id",
            "count"
        ),
        weekdays=(
            "is_weekday",
            "sum"
        ),
        weekends=(
            "is_weekend",
            "sum"
        )
    )
    .reset_index()
)


year_summary
```

```python
# ==============================================================================
# MONTH DISTRIBUTION
# ==============================================================================

print("=" * 80)
print("MONTH DISTRIBUTION")
print("=" * 80)


month_summary = (
    dim_date
    .groupby(
        [
            "year",
            "month",
            "month_name"
        ]
    )
    .agg(
        total_days=(
            "date_id",
            "count"
        ),
        weekdays=(
            "is_weekday",
            "sum"
        ),
        weekends=(
            "is_weekend",
            "sum"
        )
    )
    .reset_index()
)


month_summary.head(20)
```

```python
# ==============================================================================
# QUARTER DISTRIBUTION
# ==============================================================================

print("=" * 80)
print("QUARTER DISTRIBUTION")
print("=" * 80)


quarter_summary = (
    dim_date
    .groupby(
        [
            "year",
            "quarter_number",
            "quarter_name"
        ]
    )
    .agg(
        total_days=(
            "date_id",
            "count"
        ),
        weekdays=(
            "is_weekday",
            "sum"
        ),
        weekends=(
            "is_weekend",
            "sum"
        )
    )
    .reset_index()
)


quarter_summary
```

```python
# ==============================================================================
# WEEKEND ANALYSIS
# ==============================================================================

print("=" * 80)
print("WEEKEND ANALYSIS")
print("=" * 80)


weekend_summary = (
    dim_date[
        [
            "day_of_week_number",
            "day_name",
            "is_weekday",
            "is_weekend"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        "day_of_week_number"
    )
)


weekend_summary
```

```python
# ==============================================================================
# EXPORT DATE DIMENSION TO CSV
# ==============================================================================

os.makedirs(
    OUTPUT_DIRECTORY,
    exist_ok=True
)


def export_to_csv(dim_date):

    """
    Export date dimension data into CSV.
    """

    dim_date.to_csv(
        CSV_FILE_PATH,
        index=False
    )


    print("=" * 80)
    print("DATE DIMENSION EXPORT")
    print("=" * 80)


    print(
        "CSV file created successfully:"
    )


    print(
        CSV_FILE_PATH
    )


    return CSV_FILE_PATH


csv_file = export_to_csv(
    dim_date
)
```

```python
# ==============================================================================
# VERIFY EXPORTED FILE
# ==============================================================================

assert os.path.exists(
    csv_file
)


file_size_kb = (
    os.path.getsize(csv_file)
    / 1024
)


print("=" * 80)
print("EXPORT VERIFICATION")
print("=" * 80)


print(
    f"File exists : "
    f"{os.path.exists(csv_file)}"
)


print(
    f"File path   : "
    f"{csv_file}"
)


print(
    f"File size   : "
    f"{file_size_kb:.2f} KB"
)
```
