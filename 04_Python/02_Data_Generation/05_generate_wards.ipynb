```python
# ==============================================================================
# IMPORT REQUIRED LIBRARIES
# ==============================================================================

import os
import random
import numpy as np
import pandas as pd

from faker import Faker
from datetime import datetime, timedelta


# ==============================================================================
# INITIALIZE RANDOM GENERATORS
# ==============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

fake = Faker("en_IN")
Faker.seed(SEED)

print("Libraries imported successfully.")
print(f"Random seed: {SEED}")
```

```python
# ==============================================================================
# WARD GENERATION CONFIGURATION
# ==============================================================================

TOTAL_WARDS = 50

OUTPUT_DIRECTORY = os.path.join(
    "03_Datasets",
    "CSV_Files"
)

CSV_FILE_PATH = os.path.join(
    OUTPUT_DIRECTORY,
    "dim_wards.csv"
)

TODAY = datetime.now()

print("=" * 80)
print("WARD GENERATION CONFIGURATION")
print("=" * 80)

print(
    f"Total wards : {TOTAL_WARDS}"
)

print(
    f"Output path : {CSV_FILE_PATH}"
)
```

```python
# ==============================================================================
# WARD MASTER DATA
# ==============================================================================

WARD_TYPES = [
    "General Ward",
    "ICU",
    "NICU",
    "PICU",
    "Emergency",
    "Surgical Ward",
    "Medical Ward",
    "Pediatric Ward",
    "Maternity Ward",
    "Cardiology Ward",
    "Orthopedic Ward",
    "Oncology Ward",
    "Neurology Ward",
    "Private Ward",
    "Semi-Private Ward"
]


DEPARTMENTS = [
    "General Medicine",
    "General Surgery",
    "Cardiology",
    "Orthopedics",
    "Pediatrics",
    "Neonatology",
    "Obstetrics and Gynecology",
    "Neurology",
    "Oncology",
    "Emergency Medicine",
    "Critical Care"
]


BUILDINGS = [
    "Main Hospital",
    "North Block",
    "South Block",
    "East Block",
    "West Block"
]


WARD_STATUSES = [
    "Active",
    "Inactive",
    "Maintenance"
]


print("Ward master data loaded successfully.")
```

```python
# ==============================================================================
# WARD CODE GENERATOR
# ==============================================================================

def generate_ward_code(ward_id):

    return f"WRD{ward_id:04d}"


print(
    generate_ward_code(1)
)

print(
    generate_ward_code(50)
)
```

```python
# ==============================================================================
# WARD NAME GENERATOR
# ==============================================================================

def generate_ward_name(
    ward_type,
    ward_id
):

    return f"{ward_type} {ward_id:02d}"
```

```python
# ==============================================================================
# GENERATE WARD RECORDS
# ==============================================================================

ward_records = []


for ward_id in range(
    1,
    TOTAL_WARDS + 1
):

    # --------------------------------------------------------------------------
    # IDENTIFIERS
    # --------------------------------------------------------------------------

    ward_code = generate_ward_code(
        ward_id
    )


    # --------------------------------------------------------------------------
    # WARD TYPE
    # --------------------------------------------------------------------------

    ward_type = random.choice(
        WARD_TYPES
    )


    # --------------------------------------------------------------------------
    # DEPARTMENT
    # --------------------------------------------------------------------------

    department = random.choice(
        DEPARTMENTS
    )


    # --------------------------------------------------------------------------
    # WARD NAME
    # --------------------------------------------------------------------------

    ward_name = generate_ward_name(
        ward_type,
        ward_id
    )


    # --------------------------------------------------------------------------
    # LOCATION
    # --------------------------------------------------------------------------

    building = random.choice(
        BUILDINGS
    )

    floor_number = random.randint(
        1,
        6
    )


    # --------------------------------------------------------------------------
    # ROOM AND BED CAPACITY
    # --------------------------------------------------------------------------

    room_count = random.randint(
        5,
        30
    )

    bed_capacity = room_count * random.choice(
        [
            1,
            2,
            2,
            3,
            4
        ]
    )


    # --------------------------------------------------------------------------
    # WARD STATUS
    # --------------------------------------------------------------------------

    ward_status = random.choices(
        WARD_STATUSES,
        weights=[92, 4, 4],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # BED AVAILABILITY
    # --------------------------------------------------------------------------

    if ward_status == "Active":

        occupied_beds = random.randint(
            int(bed_capacity * 0.40),
            int(bed_capacity * 0.95)
        )

        available_beds = (
            bed_capacity
            - occupied_beds
        )

    else:

        available_beds = 0


    # --------------------------------------------------------------------------
    # OCCUPANCY RATE
    # --------------------------------------------------------------------------

    occupancy_rate = (
        (bed_capacity - available_beds)
        / bed_capacity
    ) * 100


    # --------------------------------------------------------------------------
    # AUDIT INFORMATION
    # --------------------------------------------------------------------------

    created_date = fake.date_time_between(
        start_date="-5y",
        end_date=TODAY
    )

    last_updated = fake.date_time_between(
        start_date=created_date,
        end_date=TODAY
    )


    # --------------------------------------------------------------------------
    # CREATE RECORD
    # --------------------------------------------------------------------------

    ward_records.append(
        {
            "ward_id": ward_id,
            "ward_code": ward_code,
            "ward_name": ward_name,
            "ward_type": ward_type,
            "department": department,
            "building": building,
            "floor_number": floor_number,
            "room_count": room_count,
            "bed_capacity": bed_capacity,
            "available_beds": available_beds,
            "occupancy_rate": round(
                occupancy_rate,
                2
            ),
            "ward_status": ward_status,
            "created_date": created_date,
            "last_updated": last_updated
        }
    )


print(
    f"Generated {len(ward_records):,} "
    f"ward records."
)
```

```python
# ==============================================================================
# CREATE WARD DATAFRAME
# ==============================================================================

wards_df = pd.DataFrame(
    ward_records
)


date_columns = [
    "created_date",
    "last_updated"
]


for column in date_columns:

    wards_df[column] = pd.to_datetime(
        wards_df[column]
    )


print("=" * 80)
print("WARD DATAFRAME CREATED")
print("=" * 80)

print(
    f"Rows    : {wards_df.shape[0]:,}"
)

print(
    f"Columns : {wards_df.shape[1]}"
)

wards_df.head()
```

```python
# ==============================================================================
# DATA TYPE CHECK
# ==============================================================================

print("=" * 80)
print("WARD DATA TYPES")
print("=" * 80)

print(
    wards_df.dtypes
)
```

```python
# ==============================================================================
# WARD DATA VALIDATION
# ==============================================================================

print("=" * 80)
print("WARD DATA VALIDATION")
print("=" * 80)


# ------------------------------------------------------------------------------
# 1. Row count
# ------------------------------------------------------------------------------

assert (
    len(wards_df)
    == TOTAL_WARDS
)

print(
    "✓ Row count validation passed."
)


# ------------------------------------------------------------------------------
# 2. Ward ID uniqueness
# ------------------------------------------------------------------------------

assert wards_df[
    "ward_id"
].is_unique

print(
    "✓ Ward ID uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 3. Ward code uniqueness
# ------------------------------------------------------------------------------

assert wards_df[
    "ward_code"
].is_unique

print(
    "✓ Ward code uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 4. Required fields
# ------------------------------------------------------------------------------

required_columns = [
    "ward_id",
    "ward_code",
    "ward_name",
    "ward_type",
    "department",
    "building",
    "floor_number",
    "room_count",
    "bed_capacity",
    "ward_status"
]


for column in required_columns:

    assert wards_df[
        column
    ].notna().all(), (
        f"Null values found in {column}"
    )


print(
    "✓ Required field validation passed."
)


# ------------------------------------------------------------------------------
# 5. Ward type validation
# ------------------------------------------------------------------------------

assert wards_df[
    "ward_type"
].isin(
    WARD_TYPES
).all()

print(
    "✓ Ward type validation passed."
)


# ------------------------------------------------------------------------------
# 6. Ward status validation
# ------------------------------------------------------------------------------

assert wards_df[
    "ward_status"
].isin(
    WARD_STATUSES
).all()

print(
    "✓ Ward status validation passed."
)


# ------------------------------------------------------------------------------
# 7. Capacity validation
# ------------------------------------------------------------------------------

assert (
    wards_df["bed_capacity"] > 0
).all()

assert (
    wards_df["available_beds"] >= 0
).all()

assert (
    wards_df["available_beds"]
    <=
    wards_df["bed_capacity"]
).all()

print(
    "✓ Bed capacity validation passed."
)


# ------------------------------------------------------------------------------
# 8. Occupancy validation
# ------------------------------------------------------------------------------

assert (
    wards_df["occupancy_rate"] >= 0
).all()

assert (
    wards_df["occupancy_rate"] <= 100
).all()

print(
    "✓ Occupancy rate validation passed."
)


# ------------------------------------------------------------------------------
# 9. Ward code format
# ------------------------------------------------------------------------------

assert wards_df[
    "ward_code"
].str.match(
    r"^WRD\d{4}$"
).all()

print(
    "✓ Ward code format validation passed."
)


print()

print(
    "All ward data validations passed successfully. ✓"
)
```

```python
# ==============================================================================
# EXPORT WARD DATA TO CSV
# ==============================================================================

os.makedirs(
    OUTPUT_DIRECTORY,
    exist_ok=True
)


def export_to_csv(wards_df):
    """
    Export ward dimension data into CSV.
    """

    wards_df.to_csv(
        CSV_FILE_PATH,
        index=False
    )


    print("=" * 80)
    print("WARD DATA EXPORT")
    print("=" * 80)


    print(
        "CSV file created successfully:"
    )


    print(
        CSV_FILE_PATH
    )


    return CSV_FILE_PATH


csv_file = export_to_csv(
    wards_df
)
```

```python
# ==============================================================================
# VERIFY EXPORTED FILE
# ==============================================================================

assert os.path.exists(
    csv_file
)


file_size_kb = (
    os.path.getsize(csv_file)
    / 1024
)


print("=" * 80)
print("EXPORT VERIFICATION")
print("=" * 80)


print(
    f"File exists : "
    f"{os.path.exists(csv_file)}"
)


print(
    f"File path   : "
    f"{csv_file}"
)


print(
    f"File size   : "
    f"{file_size_kb:.2f} KB"
)
```

```python
# ==============================================================================
# READ BACK EXPORTED CSV
# ==============================================================================

wards_csv_check = pd.read_csv(
    csv_file
)


print("=" * 80)
print("EXPORTED CSV VERIFICATION")
print("=" * 80)


print(
    f"Rows read from CSV    : "
    f"{len(wards_csv_check):,}"
)


print(
    f"Columns read from CSV : "
    f"{len(wards_csv_check.columns)}"
)


assert (
    len(wards_csv_check)
    == TOTAL_WARDS
)


print(
    "✓ Exported CSV row count verified."
)
```
