```python
# ==============================================================================
# IMPORT REQUIRED LIBRARIES
# ==============================================================================

import os
import random
import numpy as np
import pandas as pd

from faker import Faker
from datetime import datetime


# ==============================================================================
# INITIALIZE RANDOM GENERATORS
# ==============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

fake = Faker("en_IN")
Faker.seed(SEED)

print("Libraries imported successfully.")
print(f"Random seed: {SEED}")
```

```python
# ==============================================================================
# LAB TEST GENERATION CONFIGURATION
# ==============================================================================

TOTAL_LAB_TESTS = 250

OUTPUT_DIRECTORY = os.path.join(
    "03_Datasets",
    "CSV_Files"
)

CSV_FILE_PATH = os.path.join(
    OUTPUT_DIRECTORY,
    "dim_lab_tests.csv"
)

TODAY = datetime.now()


print("=" * 80)
print("LAB TEST GENERATION CONFIGURATION")
print("=" * 80)

print(
    f"Total lab tests : {TOTAL_LAB_TESTS}"
)

print(
    f"Output path     : {CSV_FILE_PATH}"
)
```

```python
# ==============================================================================
# LAB TEST MASTER DATA
# ==============================================================================

TEST_CATEGORIES = [
    "Hematology",
    "Biochemistry",
    "Microbiology",
    "Immunology",
    "Serology",
    "Pathology",
    "Molecular Diagnostics",
    "Endocrinology",
    "Cardiology",
    "Coagulation",
    "Urinalysis",
    "Toxicology"
]


TEST_SUBCATEGORIES = [
    "Routine",
    "Specialized",
    "Screening",
    "Diagnostic",
    "Monitoring",
    "Emergency",
    "Preventive"
]


SPECIMEN_TYPES = [
    "Blood",
    "Serum",
    "Plasma",
    "Urine",
    "Stool",
    "Sputum",
    "Swab",
    "CSF",
    "Tissue",
    "Pleural Fluid",
    "Body Fluid"
]


DEPARTMENTS = [
    "Laboratory Medicine",
    "Hematology",
    "Biochemistry",
    "Microbiology",
    "Pathology",
    "Cardiology",
    "Endocrinology",
    "Emergency Medicine",
    "Oncology",
    "Nephrology",
    "Pulmonology"
]


RESULT_TYPES = [
    "Numeric",
    "Text",
    "Positive/Negative",
    "Reactive/Non-Reactive",
    "Detected/Not Detected"
]


TEST_STATUSES = [
    "Active",
    "Inactive",
    "Discontinued"
]


print("Lab test master data loaded successfully.")
```

```python
# ==============================================================================
# LAB TEST NAME MASTER DATA
# ==============================================================================

TEST_NAMES = [
    "Complete Blood Count",
    "Hemoglobin",
    "Hematocrit",
    "Red Blood Cell Count",
    "White Blood Cell Count",
    "Platelet Count",
    "Mean Corpuscular Volume",
    "Mean Corpuscular Hemoglobin",
    "Mean Corpuscular Hemoglobin Concentration",
    "Erythrocyte Sedimentation Rate",
    "C-Reactive Protein",
    "Blood Glucose Fasting",
    "Blood Glucose Random",
    "HbA1c",
    "Lipid Profile",
    "Total Cholesterol",
    "HDL Cholesterol",
    "LDL Cholesterol",
    "Triglycerides",
    "Liver Function Test",
    "ALT",
    "AST",
    "Alkaline Phosphatase",
    "Bilirubin Total",
    "Bilirubin Direct",
    "Total Protein",
    "Albumin",
    "Kidney Function Test",
    "Serum Creatinine",
    "Blood Urea Nitrogen",
    "Uric Acid",
    "Serum Sodium",
    "Serum Potassium",
    "Serum Calcium",
    "Thyroid Stimulating Hormone",
    "Free T3",
    "Free T4",
    "Vitamin B12",
    "Vitamin D",
    "Iron Studies",
    "Serum Ferritin",
    "Serum Iron",
    "Total Iron Binding Capacity",
    "Prothrombin Time",
    "INR",
    "Activated Partial Thromboplastin Time",
    "D-Dimer",
    "Blood Group",
    "Urine Routine Examination",
    "Urine Culture",
    "Stool Routine Examination",
    "Stool Occult Blood",
    "Blood Culture",
    "Urine Protein",
    "Urine Glucose",
    "Urine Ketones",
    "Pregnancy Test",
    "HIV Test",
    "Hepatitis B Surface Antigen",
    "Hepatitis C Antibody",
    "Dengue NS1 Antigen",
    "Dengue IgM",
    "Dengue IgG",
    "Malaria Parasite Test",
    "COVID-19 PCR",
    "Influenza A/B Test",
    "Tuberculosis PCR",
    "Troponin I",
    "Troponin T",
    "CK-MB",
    "BNP",
    "NT-proBNP",
    "Amylase",
    "Lipase",
    "Lactate",
    "Procalcitonin",
    "Blood Gas Analysis",
    "Arterial Blood Gas",
    "Venous Blood Gas"
]


print(
    f"Available test names: {len(TEST_NAMES)}"
)
```

```python
# ==============================================================================
# UNIT AND REFERENCE RANGE MASTER DATA
# ==============================================================================

UNITS = [
    "mg/dL",
    "g/dL",
    "IU/L",
    "U/L",
    "mmol/L",
    "mEq/L",
    "ng/mL",
    "pg/mL",
    "µg/mL",
    "cells/µL",
    "%",
    "seconds",
    "ratio",
    "mIU/L",
    "Negative",
    "Positive",
    "Not Detected"
]


REFERENCE_RANGES = [
    "3.5 - 5.5",
    "4.0 - 6.0",
    "10 - 40",
    "12 - 16",
    "60 - 100",
    "70 - 110",
    "0.6 - 1.2",
    "0.5 - 1.0",
    "0 - 5",
    "1 - 10",
    "0 - 1",
    "Negative",
    "Positive",
    "Not Detected"
]


print("Unit and reference range data loaded successfully.")
```

```python
# ==============================================================================
# LAB TEST CODE GENERATOR
# ==============================================================================

def generate_lab_test_code(lab_test_id):

    return f"LAB{lab_test_id:05d}"


print(
    generate_lab_test_code(1)
)

print(
    generate_lab_test_code(
        TOTAL_LAB_TESTS
    )
)
```

```python
# ==============================================================================
# GENERATE LAB TEST RECORDS
# ==============================================================================

lab_test_records = []


for lab_test_id in range(
    1,
    TOTAL_LAB_TESTS + 1
):

    # --------------------------------------------------------------------------
    # IDENTIFIERS
    # --------------------------------------------------------------------------

    lab_test_code = generate_lab_test_code(
        lab_test_id
    )


    # --------------------------------------------------------------------------
    # TEST NAME
    # --------------------------------------------------------------------------

    base_test_name = random.choice(
        TEST_NAMES
    )

    test_name = (
        base_test_name
        + " "
        + random.choice(
            [
                "Panel",
                "Test",
                "Analysis",
                "Screen",
                "Profile"
            ]
        )
    )


    # --------------------------------------------------------------------------
    # CATEGORY
    # --------------------------------------------------------------------------

    test_category = random.choice(
        TEST_CATEGORIES
    )


    # --------------------------------------------------------------------------
    # SUBCATEGORY
    # --------------------------------------------------------------------------

    test_subcategory = random.choice(
        TEST_SUBCATEGORIES
    )


    # --------------------------------------------------------------------------
    # SPECIMEN
    # --------------------------------------------------------------------------

    specimen_type = random.choice(
        SPECIMEN_TYPES
    )


    # --------------------------------------------------------------------------
    # DEPARTMENT
    # --------------------------------------------------------------------------

    department = random.choice(
        DEPARTMENTS
    )


    # --------------------------------------------------------------------------
    # RESULT TYPE
    # --------------------------------------------------------------------------

    result_type = random.choice(
        RESULT_TYPES
    )


    # --------------------------------------------------------------------------
    # UNIT
    # --------------------------------------------------------------------------

    if result_type in [
        "Positive/Negative",
        "Reactive/Non-Reactive",
        "Detected/Not Detected"
    ]:

        unit_of_measure = "N/A"

    else:

        unit_of_measure = random.choice(
            UNITS[:14]
        )


    # --------------------------------------------------------------------------
    # REFERENCE RANGE
    # --------------------------------------------------------------------------

    reference_range = random.choice(
        REFERENCE_RANGES
    )


    # --------------------------------------------------------------------------
    # CRITICAL LIMITS
    # --------------------------------------------------------------------------

    critical_low = round(
        random.uniform(
            0,
            20
        ),
        2
    )

    critical_high = round(
        random.uniform(
            50,
            500
        ),
        2
    )


    # --------------------------------------------------------------------------
    # TEST COST
    # --------------------------------------------------------------------------

    test_cost = round(
        random.uniform(
            100,
            5000
        ),
        2
    )


    # --------------------------------------------------------------------------
    # TURNAROUND TIME
    # --------------------------------------------------------------------------

    turnaround_time_hours = random.choice(
        [
            1,
            2,
            4,
            6,
            8,
            12,
            24,
            48,
            72
        ]
    )


    # --------------------------------------------------------------------------
    # TEST STATUS
    # --------------------------------------------------------------------------

    test_status = random.choices(
        TEST_STATUSES,
        weights=[
            94,
            4,
            2
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # AUDIT INFORMATION
    # --------------------------------------------------------------------------

    created_date = fake.date_time_between(
        start_date="-5y",
        end_date=TODAY
    )

    last_updated = fake.date_time_between(
        start_date=created_date,
        end_date=TODAY
    )


    # --------------------------------------------------------------------------
    # CREATE RECORD
    # --------------------------------------------------------------------------

    lab_test_records.append(
        {
            "lab_test_id": lab_test_id,
            "lab_test_code": lab_test_code,
            "test_name": test_name,
            "test_category": test_category,
            "test_subcategory": test_subcategory,
            "specimen_type": specimen_type,
            "department": department,
            "result_type": result_type,
            "unit_of_measure": unit_of_measure,
            "reference_range": reference_range,
            "critical_low": critical_low,
            "critical_high": critical_high,
            "test_cost": test_cost,
            "turnaround_time_hours": turnaround_time_hours,
            "test_status": test_status,
            "created_date": created_date,
            "last_updated": last_updated
        }
    )


print(
    f"Generated {len(lab_test_records):,} "
    f"lab test records."
)
```

```python
# ==============================================================================
# CREATE LAB TEST DATAFRAME
# ==============================================================================

lab_tests_df = pd.DataFrame(
    lab_test_records
)


date_columns = [
    "created_date",
    "last_updated"
]


for column in date_columns:

    lab_tests_df[column] = pd.to_datetime(
        lab_tests_df[column]
    )


print("=" * 80)
print("LAB TEST DATAFRAME CREATED")
print("=" * 80)

print(
    f"Rows    : {lab_tests_df.shape[0]:,}"
)

print(
    f"Columns : {lab_tests_df.shape[1]}"
)

lab_tests_df.head()
```

```python
# ==============================================================================
# DATA TYPE CHECK
# ==============================================================================

print("=" * 80)
print("LAB TEST DATA TYPES")
print("=" * 80)

print(
    lab_tests_df.dtypes
)
```

```python
# ==============================================================================
# LAB TEST DATA VALIDATION
# ==============================================================================

print("=" * 80)
print("LAB TEST DATA VALIDATION")
print("=" * 80)


# ------------------------------------------------------------------------------
# 1. Row count
# ------------------------------------------------------------------------------

assert (
    len(lab_tests_df)
    == TOTAL_LAB_TESTS
)

print(
    "✓ Row count validation passed."
)


# ------------------------------------------------------------------------------
# 2. Lab test ID uniqueness
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "lab_test_id"
].is_unique

print(
    "✓ Lab test ID uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 3. Lab test code uniqueness
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "lab_test_code"
].is_unique

print(
    "✓ Lab test code uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 4. Test name uniqueness
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "test_name"
].is_unique

print(
    "✓ Test name uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 5. Required fields
# ------------------------------------------------------------------------------

required_columns = [
    "lab_test_id",
    "lab_test_code",
    "test_name",
    "test_category",
    "test_subcategory",
    "specimen_type",
    "department",
    "result_type",
    "unit_of_measure",
    "reference_range",
    "test_cost",
    "turnaround_time_hours",
    "test_status"
]


for column in required_columns:

    assert lab_tests_df[
        column
    ].notna().all(), (
        f"Null values found in {column}"
    )


print(
    "✓ Required field validation passed."
)


# ------------------------------------------------------------------------------
# 6. Category validation
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "test_category"
].isin(
    TEST_CATEGORIES
).all()

print(
    "✓ Test category validation passed."
)


# ------------------------------------------------------------------------------
# 7. Subcategory validation
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "test_subcategory"
].isin(
    TEST_SUBCATEGORIES
).all()

print(
    "✓ Test subcategory validation passed."
)


# ------------------------------------------------------------------------------
# 8. Specimen validation
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "specimen_type"
].isin(
    SPECIMEN_TYPES
).all()

print(
    "✓ Specimen type validation passed."
)


# ------------------------------------------------------------------------------
# 9. Department validation
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "department"
].isin(
    DEPARTMENTS
).all()

print(
    "✓ Department validation passed."
)


# ------------------------------------------------------------------------------
# 10. Result type validation
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "result_type"
].isin(
    RESULT_TYPES
).all()

print(
    "✓ Result type validation passed."
)


# ------------------------------------------------------------------------------
# 11. Financial validation
# ------------------------------------------------------------------------------

assert (
    lab_tests_df[
        "test_cost"
    ] > 0
).all()

print(
    "✓ Test cost validation passed."
)


# ------------------------------------------------------------------------------
# 12. Turnaround time validation
# ------------------------------------------------------------------------------

assert (
    lab_tests_df[
        "turnaround_time_hours"
    ] > 0
).all()

print(
    "✓ Turnaround time validation passed."
)


# ------------------------------------------------------------------------------
# 13. Critical range validation
# ------------------------------------------------------------------------------

assert (
    lab_tests_df[
        "critical_low"
    ] >= 0
).all()


assert (
    lab_tests_df[
        "critical_high"
    ]
    >
    lab_tests_df[
        "critical_low"
    ]
).all()

print(
    "✓ Critical range validation passed."
)


# ------------------------------------------------------------------------------
# 14. Test status validation
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "test_status"
].isin(
    TEST_STATUSES
).all()

print(
    "✓ Test status validation passed."
)


# ------------------------------------------------------------------------------
# 15. Lab test code format
# ------------------------------------------------------------------------------

assert lab_tests_df[
    "lab_test_code"
].str.match(
    r"^LAB\d{5}$"
).all()

print(
    "✓ Lab test code format validation passed."
)


print()

print(
    "All lab test data validations passed successfully. ✓"
)
```

```python
# ==============================================================================
# LAB TEST CATEGORY COST SUMMARY
# ==============================================================================

print("=" * 80)
print("LAB TEST CATEGORY COST SUMMARY")
print("=" * 80)


category_cost_summary = (
    lab_tests_df
    .groupby(
        "test_category"
    )
    .agg(
        test_count=(
            "lab_test_id",
            "count"
        ),
        average_test_cost=(
            "test_cost",
            "mean"
        ),
        minimum_test_cost=(
            "test_cost",
            "min"
        ),
        maximum_test_cost=(
            "test_cost",
            "max"
        ),
        average_turnaround_hours=(
            "turnaround_time_hours",
            "mean"
        )
    )
    .reset_index()
)


category_cost_summary[
    "average_test_cost"
] = category_cost_summary[
    "average_test_cost"
].round(2)


category_cost_summary[
    "average_turnaround_hours"
] = category_cost_summary[
    "average_turnaround_hours"
].round(2)


category_cost_summary
```

```python
# ==============================================================================
# EXPORT LAB TEST DATA TO CSV
# ==============================================================================

os.makedirs(
    OUTPUT_DIRECTORY,
    exist_ok=True
)


def export_to_csv(lab_tests_df):
    """
    Export lab test dimension data into CSV.
    """

    lab_tests_df.to_csv(
        CSV_FILE_PATH,
        index=False
    )


    print("=" * 80)
    print("LAB TEST DATA EXPORT")
    print("=" * 80)


    print(
        "CSV file created successfully:"
    )


    print(
        CSV_FILE_PATH
    )


    return CSV_FILE_PATH


csv_file = export_to_csv(
    lab_tests_df
)
```
