# 🏥 Hospital Analytics Portfolio Project
## Insurance Dimension Data Generation

This notebook generates synthetic insurance master data
for the Hospital Analytics Portfolio Project.

Output:
    dim_insurance.csv

The insurance dimension contains insurance provider,
policy, coverage, premium, deductible, validity,
and status information.
```

````python
# ==============================================================================
# IMPORT REQUIRED LIBRARIES
# ==============================================================================

import os
import random
import numpy as np
import pandas as pd

from faker import Faker
from datetime import datetime, timedelta


# ==============================================================================
# INITIALIZE RANDOM GENERATORS
# ==============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

fake = Faker("en_IN")
Faker.seed(SEED)

print("Libraries imported successfully.")
print(f"Random seed: {SEED}")
````

````python
# ==============================================================================
# INSURANCE DATA CONFIGURATION
# ==============================================================================

TOTAL_INSURANCE_POLICIES = 1000

CSV_FILE_NAME = "dim_insurance.csv"

OUTPUT_DIR = "03_Datasets/CSV_Files"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

CSV_FILE_PATH = os.path.join(
    OUTPUT_DIR,
    CSV_FILE_NAME
)

print("=" * 80)
print("INSURANCE DATA CONFIGURATION")
print("=" * 80)
print(f"Total insurance policies : {TOTAL_INSURANCE_POLICIES:,}")
print(f"Output file              : {CSV_FILE_PATH}")
````

## Insurance Dimension Schema

The insurance dimension stores policy-level master information.

The following fields are generated:

- insurance_id
- insurance_code
- insurance_provider
- policy_number
- policy_type
- plan_name
- coverage_amount
- deductible_amount
- premium_amount
- policy_start_date
- policy_end_date
- insurance_status
- created_date
- last_updated
```

---

````python
# ==============================================================================
# MASTER INSURANCE VALUES
# ==============================================================================

INSURANCE_PROVIDERS = [
    "Star Health and Allied Insurance",
    "HDFC ERGO General Insurance",
    "ICICI Lombard General Insurance",
    "Niva Bupa Health Insurance",
    "Care Health Insurance",
    "Aditya Birla Health Insurance",
    "Bajaj Allianz General Insurance",
    "ManipalCigna Health Insurance",
    "New India Assurance",
    "United India Insurance"
]

POLICY_TYPES = [
    "Individual",
    "Family Floater",
    "Senior Citizen",
    "Corporate"
]

PLAN_NAMES = [
    "Basic Health Plan",
    "Comprehensive Health Plan",
    "Premium Health Plan",
    "Family Care Plan",
    "Senior Care Plan",
    "Corporate Health Plan"
]

INSURANCE_STATUSES = [
    "Active",
    "Expired",
    "Cancelled"
]

print("Insurance master values configured successfully.")
````

````python
# ==============================================================================
# DATE CONFIGURATION
# ==============================================================================

TODAY = datetime(2026, 8, 4)

MIN_POLICY_START_DATE = datetime(2020, 1, 1)
MAX_POLICY_START_DATE = TODAY

print("=" * 80)
print("DATE CONFIGURATION")
print("=" * 80)

print(
    f"Policy start date range : "
    f"{MIN_POLICY_START_DATE.date()} "
    f"to "
    f"{MAX_POLICY_START_DATE.date()}"
)

print(
    f"Reference date          : {TODAY.date()}"
)
````

````python
# ==============================================================================
# HELPER FUNCTIONS
# ==============================================================================

def generate_insurance_code(insurance_id):
    """
    Generate a unique insurance business code.
    """
    return f"INS{insurance_id:06d}"


def generate_policy_number(insurance_id):
    """
    Generate a unique policy number.
    """
    return f"POL{insurance_id:08d}"


def generate_policy_dates():
    """
    Generate policy start and end dates.
    Policies are generally created for one-year periods.
    """

    start_date = fake.date_between(
        start_date=MIN_POLICY_START_DATE.date(),
        end_date=MAX_POLICY_START_DATE.date()
    )

    start_date = datetime.combine(
        start_date,
        datetime.min.time()
    )

    end_date = start_date + timedelta(
        days=random.choice([364, 365, 730])
    )

    return start_date, end_date


print("Helper functions created successfully.")
````

````python
# ==============================================================================
# GENERATE INSURANCE RECORDS
# ==============================================================================

insurance_records = []

for insurance_id in range(
    1,
    TOTAL_INSURANCE_POLICIES + 1
):

    # --------------------------------------------------------------------------
    # IDENTIFIERS
    # --------------------------------------------------------------------------

    insurance_code = generate_insurance_code(
        insurance_id
    )

    policy_number = generate_policy_number(
        insurance_id
    )

    # --------------------------------------------------------------------------
    # PROVIDER AND PLAN
    # --------------------------------------------------------------------------

    insurance_provider = random.choice(
        INSURANCE_PROVIDERS
    )

    policy_type = random.choice(
        POLICY_TYPES
    )

    plan_name = random.choice(
        PLAN_NAMES
    )

    # --------------------------------------------------------------------------
    # COVERAGE AND FINANCIAL DETAILS
    # --------------------------------------------------------------------------

    coverage_amount = random.choice(
        [
            200000,
            300000,
            500000,
            750000,
            1000000,
            1500000,
            2000000,
            2500000,
            5000000
        ]
    )

    deductible_amount = random.choice(
        [
            0,
            5000,
            10000,
            15000,
            25000,
            50000
        ]
    )

    premium_amount = random.choice(
        [
            5000,
            7500,
            10000,
            12500,
            15000,
            20000,
            25000,
            30000,
            40000,
            50000
        ]
    )

    # --------------------------------------------------------------------------
    # POLICY DATES
    # --------------------------------------------------------------------------

    policy_start_date, policy_end_date = generate_policy_dates()

    # --------------------------------------------------------------------------
    # POLICY STATUS
    # --------------------------------------------------------------------------

    if policy_end_date < TODAY:

        insurance_status = random.choices(
            ["Expired", "Cancelled"],
            weights=[95, 5],
            k=1
        )[0]

    else:

        insurance_status = "Active"

    # --------------------------------------------------------------------------
    # AUDIT INFORMATION
    # --------------------------------------------------------------------------

    created_date = policy_start_date

    last_updated = fake.date_time_between(
        start_date=created_date,
        end_date=min(
            policy_end_date,
            TODAY
        )
    )

    # --------------------------------------------------------------------------
    # CREATE RECORD
    # --------------------------------------------------------------------------

    insurance_records.append(
        {
            "insurance_id": insurance_id,
            "insurance_code": insurance_code,
            "insurance_provider": insurance_provider,
            "policy_number": policy_number,
            "policy_type": policy_type,
            "plan_name": plan_name,
            "coverage_amount": coverage_amount,
            "deductible_amount": deductible_amount,
            "premium_amount": premium_amount,
            "policy_start_date": policy_start_date,
            "policy_end_date": policy_end_date,
            "insurance_status": insurance_status,
            "created_date": created_date,
            "last_updated": last_updated
        }
    )

print(
    f"Generated {len(insurance_records):,} "
    f"insurance records."
)
````

````python
# ==============================================================================
# CREATE INSURANCE DATAFRAME
# ==============================================================================

insurance_df = pd.DataFrame(
    insurance_records
)

date_columns = [
    "policy_start_date",
    "policy_end_date",
    "created_date",
    "last_updated"
]

for column in date_columns:

    insurance_df[column] = pd.to_datetime(
        insurance_df[column]
    )

print("=" * 80)
print("INSURANCE DATAFRAME CREATED")
print("=" * 80)

print(
    f"Rows    : {insurance_df.shape[0]:,}"
)

print(
    f"Columns : {insurance_df.shape[1]}"
)

insurance_df.head()
````

````python
# ==============================================================================
# DATA TYPE CHECK
# ==============================================================================

print("=" * 80)
print("DATA TYPES")
print("=" * 80)

print(
    insurance_df.dtypes
)
````

````python
# ==============================================================================
# INSURANCE DATA VALIDATION
# ==============================================================================

print("=" * 80)
print("INSURANCE DATA VALIDATION")
print("=" * 80)


# ------------------------------------------------------------------------------
# 1. Row count
# ------------------------------------------------------------------------------

assert (
    len(insurance_df)
    == TOTAL_INSURANCE_POLICIES
)

print("✓ Row count validation passed.")


# ------------------------------------------------------------------------------
# 2. Insurance ID uniqueness
# ------------------------------------------------------------------------------

assert insurance_df[
    "insurance_id"
].is_unique

print(
    "✓ Insurance ID uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 3. Insurance code uniqueness
# ------------------------------------------------------------------------------

assert insurance_df[
    "insurance_code"
].is_unique

print(
    "✓ Insurance code uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 4. Policy number uniqueness
# ------------------------------------------------------------------------------

assert insurance_df[
    "policy_number"
].is_unique

print(
    "✓ Policy number uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 5. Required fields
# ------------------------------------------------------------------------------

required_columns = [
    "insurance_id",
    "insurance_code",
    "insurance_provider",
    "policy_number",
    "policy_type",
    "plan_name",
    "coverage_amount",
    "premium_amount",
    "policy_start_date",
    "policy_end_date"
]

for column in required_columns:

    assert insurance_df[
        column
    ].notna().all(), (
        f"Null values found in {column}"
    )

print(
    "✓ Required field validation passed."
)


# ------------------------------------------------------------------------------
# 6. Policy type validation
# ------------------------------------------------------------------------------

assert insurance_df[
    "policy_type"
].isin(
    POLICY_TYPES
).all()

print(
    "✓ Policy type validation passed."
)


# ------------------------------------------------------------------------------
# 7. Status validation
# ------------------------------------------------------------------------------

assert insurance_df[
    "insurance_status"
].isin(
    INSURANCE_STATUSES
).all()

print(
    "✓ Insurance status validation passed."
)


# ------------------------------------------------------------------------------
# 8. Financial validation
# ------------------------------------------------------------------------------

assert (
    insurance_df["coverage_amount"] > 0
).all()

assert (
    insurance_df["premium_amount"] >= 0
).all()

assert (
    insurance_df["deductible_amount"] >= 0
).all()

print(
    "✓ Financial validation passed."
)


# ------------------------------------------------------------------------------
# 9. Policy date validation
# ------------------------------------------------------------------------------

assert (
    insurance_df["policy_end_date"]
    >= insurance_df["policy_start_date"]
).all()

print(
    "✓ Policy date validation passed."
)


# ------------------------------------------------------------------------------
# 10. Insurance code format
# ------------------------------------------------------------------------------

assert insurance_df[
    "insurance_code"
].str.match(
    r"^INS\d{6}$"
).all()

print(
    "✓ Insurance code format validation passed."
)


print()
print(
    "All insurance data validations passed successfully. ✓"
)
````

````python
# ==============================================================================
# CHECK DUPLICATE RECORDS
# ==============================================================================

duplicate_count = insurance_df.duplicated().sum()

print("=" * 80)
print("DUPLICATE CHECK")
print("=" * 80)

print(
    f"Duplicate records: {duplicate_count}"
)

assert duplicate_count == 0

print(
    "✓ No duplicate records found."
)
````

````python
# ==============================================================================
# EXPORT INSURANCE DATA TO CSV
# ==============================================================================

def export_to_csv(insurance_df):
    """
    Export insurance dimension data into CSV.
    """

    insurance_df.to_csv(
        CSV_FILE_PATH,
        index=False
    )

    print("=" * 80)
    print("INSURANCE DATA EXPORT")
    print("=" * 80)

    print(
        "CSV file created successfully:"
    )

    print(
        CSV_FILE_PATH
    )

    return CSV_FILE_PATH


csv_file = export_to_csv(
    insurance_df
)
````

````python
# ==============================================================================
# VERIFY EXPORTED FILE
# ==============================================================================

assert os.path.exists(
    csv_file
)

file_size_kb = (
    os.path.getsize(csv_file)
    / 1024
)

print("=" * 80)
print("EXPORT VERIFICATION")
print("=" * 80)

print(
    f"File exists : {os.path.exists(csv_file)}"
)

print(
    f"File path   : {csv_file}"
)

print(
    f"File size   : {file_size_kb:.2f} KB"
)
````

````python
# ==============================================================================
# READ BACK EXPORTED CSV
# ==============================================================================

insurance_csv_check = pd.read_csv(
    csv_file
)

print("=" * 80)
print("EXPORTED CSV VERIFICATION")
print("=" * 80)

print(
    f"Rows read from CSV    : "
    f"{len(insurance_csv_check):,}"
)

print(
    f"Columns read from CSV : "
    f"{len(insurance_csv_check.columns)}"
)

assert (
    len(insurance_csv_check)
    == TOTAL_INSURANCE_POLICIES
)

print(
    "✓ Exported CSV row count verified."
)
````
