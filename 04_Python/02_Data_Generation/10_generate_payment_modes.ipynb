```python
# ==============================================================================
# IMPORT REQUIRED LIBRARIES
# ==============================================================================

import os
import random
import numpy as np
import pandas as pd

from faker import Faker
from datetime import datetime


# ==============================================================================
# INITIALIZE RANDOM GENERATORS
# ==============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

fake = Faker("en_IN")
Faker.seed(SEED)

print("Libraries imported successfully.")
print(f"Random seed: {SEED}")
```

```python
# ==============================================================================
# PAYMENT MASTER DATA
# ==============================================================================

PAYMENT_CATEGORIES = [
    "Cash",
    "Card",
    "Banking",
    "Digital Wallet",
    "UPI",
    "Insurance",
    "Government",
    "Corporate",
    "Other"
]


PAYMENT_METHOD_TYPES = [
    "Cash",
    "Debit Card",
    "Credit Card",
    "UPI",
    "Net Banking",
    "Mobile Wallet",
    "Bank Transfer",
    "Cheque",
    "Demand Draft",
    "Insurance Claim",
    "Corporate Credit",
    "Government Scheme"
]


TRANSACTION_TYPES = [
    "Payment",
    "Refund",
    "Adjustment",
    "Advance Payment"
]


CURRENCIES = [
    "INR"
]


PAYMENT_MODE_STATUSES = [
    "Active",
    "Inactive",
    "Deprecated"
]


print("Payment master data loaded successfully.")
```

```python
# ==============================================================================
# PAYMENT MODE NAME MASTER DATA
# ==============================================================================

PAYMENT_MODE_NAMES = [
    "Cash",
    "Visa Debit Card",
    "Mastercard Debit Card",
    "RuPay Debit Card",
    "Visa Credit Card",
    "Mastercard Credit Card",
    "RuPay Credit Card",
    "UPI",
    "Google Pay",
    "PhonePe",
    "Paytm",
    "Net Banking",
    "NEFT",
    "RTGS",
    "IMPS",
    "Bank Transfer",
    "Cheque",
    "Demand Draft",
    "Insurance Claim",
    "Government Health Scheme",
    "Corporate Credit",
    "Hospital Wallet",
    "Online Payment Gateway",
    "Mobile Wallet",
    "Other"
]


print(
    f"Available payment modes: "
    f"{len(PAYMENT_MODE_NAMES)}"
)
```

```python
# ==============================================================================
# PAYMENT MODE CODE GENERATOR
# ==============================================================================

# Total number of payment modes
TOTAL_PAYMENT_MODES = 25


def generate_payment_mode_code(payment_mode_id):

    return f"PAY{payment_mode_id:05d}"


print("=" * 80)
print("PAYMENT MODE CODE GENERATOR")
print("=" * 80)

print(
    f"First payment mode code : "
    f"{generate_payment_mode_code(1)}"
)

print(
    f"Last payment mode code  : "
    f"{generate_payment_mode_code(TOTAL_PAYMENT_MODES)}"
)
```

```python
# ==============================================================================
# GENERATE PAYMENT MODE RECORDS
# ==============================================================================

# ------------------------------------------------------------------------------
# SAFETY CONFIGURATION
# ------------------------------------------------------------------------------

from datetime import datetime

TODAY = datetime.now()

TOTAL_PAYMENT_MODES = 25


# ------------------------------------------------------------------------------
# VERIFY REQUIRED VARIABLES
# ------------------------------------------------------------------------------

required_variables = [
    "PAYMENT_MODE_NAMES",
    "TRANSACTION_TYPES",
    "PAYMENT_MODE_STATUSES"
]

missing_variables = [
    variable
    for variable in required_variables
    if variable not in globals()
]

if missing_variables:

    raise NameError(
        "The following required variables are missing: "
        + ", ".join(missing_variables)
        + ". Please run the master-data cell first."
    )


# ==============================================================================
# GENERATE PAYMENT MODE RECORDS
# ==============================================================================

payment_mode_records = []


for payment_mode_id in range(
    1,
    TOTAL_PAYMENT_MODES + 1
):

    # --------------------------------------------------------------------------
    # IDENTIFIERS
    # --------------------------------------------------------------------------

    payment_mode_code = generate_payment_mode_code(
        payment_mode_id
    )


    # --------------------------------------------------------------------------
    # PAYMENT MODE NAME
    # --------------------------------------------------------------------------

    payment_mode_name = PAYMENT_MODE_NAMES[
        payment_mode_id - 1
    ]


    # --------------------------------------------------------------------------
    # PAYMENT CATEGORY
    # --------------------------------------------------------------------------

    if payment_mode_name == "Cash":

        payment_category = "Cash"

    elif "Card" in payment_mode_name:

        payment_category = "Card"

    elif payment_mode_name in [
        "UPI",
        "Google Pay",
        "PhonePe",
        "Paytm"
    ]:

        payment_category = "UPI"

    elif payment_mode_name in [
        "Net Banking",
        "NEFT",
        "RTGS",
        "IMPS",
        "Bank Transfer"
    ]:

        payment_category = "Banking"

    elif payment_mode_name == "Insurance Claim":

        payment_category = "Insurance"

    elif payment_mode_name == "Government Health Scheme":

        payment_category = "Government"

    elif payment_mode_name == "Corporate Credit":

        payment_category = "Corporate"

    elif payment_mode_name in [
        "Hospital Wallet",
        "Mobile Wallet"
    ]:

        payment_category = "Digital Wallet"

    else:

        payment_category = "Other"


    # --------------------------------------------------------------------------
    # PAYMENT METHOD TYPE
    # --------------------------------------------------------------------------

    payment_method_type = payment_mode_name


    # --------------------------------------------------------------------------
    # PROVIDER NAME
    # --------------------------------------------------------------------------

    if payment_mode_name == "Google Pay":

        provider_name = "Google Pay"

    elif payment_mode_name == "PhonePe":

        provider_name = "PhonePe"

    elif payment_mode_name == "Paytm":

        provider_name = "Paytm"

    elif "Visa" in payment_mode_name:

        provider_name = "Visa"

    elif "Mastercard" in payment_mode_name:

        provider_name = "Mastercard"

    elif "RuPay" in payment_mode_name:

        provider_name = "RuPay"

    elif payment_mode_name == "Insurance Claim":

        provider_name = "Insurance Provider"

    elif payment_mode_name == "Government Health Scheme":

        provider_name = "Government Health Authority"

    elif payment_mode_name == "Online Payment Gateway":

        provider_name = "Payment Gateway"

    else:

        provider_name = "Hospital"


    # --------------------------------------------------------------------------
    # TRANSACTION TYPE
    # --------------------------------------------------------------------------

    transaction_type = random.choice(
        TRANSACTION_TYPES
    )


    # --------------------------------------------------------------------------
    # CURRENCY
    # --------------------------------------------------------------------------

    currency = "INR"


    # --------------------------------------------------------------------------
    # PROCESSING FEE
    # --------------------------------------------------------------------------

    if payment_category == "Cash":

        processing_fee_percent = 0.00

    elif payment_category == "UPI":

        processing_fee_percent = round(
            random.uniform(
                0.00,
                1.00
            ),
            2
        )

    elif payment_category == "Card":

        processing_fee_percent = round(
            random.uniform(
                1.00,
                3.00
            ),
            2
        )

    elif payment_category == "Banking":

        processing_fee_percent = round(
            random.uniform(
                0.00,
                1.50
            ),
            2
        )

    elif payment_category == "Digital Wallet":

        processing_fee_percent = round(
            random.uniform(
                0.50,
                2.00
            ),
            2
        )

    else:

        processing_fee_percent = round(
            random.uniform(
                0.00,
                1.00
            ),
            2
        )


    # --------------------------------------------------------------------------
    # SETTLEMENT DAYS
    # --------------------------------------------------------------------------

    if payment_category == "Cash":

        settlement_days = 0

    elif payment_category in [
        "UPI",
        "Card"
    ]:

        settlement_days = random.choice(
            [
                0,
                1,
                2
            ]
        )

    elif payment_category == "Banking":

        settlement_days = random.choice(
            [
                1,
                2,
                3
            ]
        )

    elif payment_category == "Insurance":

        settlement_days = random.choice(
            [
                7,
                15,
                30,
                45
            ]
        )

    elif payment_category == "Government":

        settlement_days = random.choice(
            [
                15,
                30,
                45,
                60
            ]
        )

    elif payment_category == "Corporate":

        settlement_days = random.choice(
            [
                15,
                30,
                45
            ]
        )

    else:

        settlement_days = random.choice(
            [
                1,
                2,
                3,
                7
            ]
        )


    # --------------------------------------------------------------------------
    # ONLINE PAYMENT
    # --------------------------------------------------------------------------

    online_payment = payment_category in [
        "Card",
        "UPI",
        "Banking",
        "Digital Wallet"
    ]


    # --------------------------------------------------------------------------
    # REFERENCE NUMBER REQUIREMENT
    # --------------------------------------------------------------------------

    requires_reference_number = (
        payment_category != "Cash"
    )


    # --------------------------------------------------------------------------
    # STATUS
    # --------------------------------------------------------------------------

    payment_mode_status = random.choices(
        PAYMENT_MODE_STATUSES,
        weights=[
            94,
            4,
            2
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # AUDIT INFORMATION
    # --------------------------------------------------------------------------

    created_date = fake.date_time_between(
        start_date="-5y",
        end_date=TODAY
    )

    last_updated = fake.date_time_between(
        start_date=created_date,
        end_date=TODAY
    )


    # --------------------------------------------------------------------------
    # CREATE RECORD
    # --------------------------------------------------------------------------

    payment_mode_records.append(
        {
            "payment_mode_id": payment_mode_id,
            "payment_mode_code": payment_mode_code,
            "payment_mode_name": payment_mode_name,
            "payment_category": payment_category,
            "payment_method_type": payment_method_type,
            "provider_name": provider_name,
            "transaction_type": transaction_type,
            "currency": currency,
            "processing_fee_percent": processing_fee_percent,
            "settlement_days": settlement_days,
            "online_payment": online_payment,
            "requires_reference_number": requires_reference_number,
            "payment_mode_status": payment_mode_status,
            "created_date": created_date,
            "last_updated": last_updated
        }
    )


# ==============================================================================
# GENERATION RESULT
# ==============================================================================

print("=" * 80)
print("PAYMENT MODE RECORD GENERATION")
print("=" * 80)

print(
    f"Generated {len(payment_mode_records):,} "
    f"payment mode records."
)

print(
    f"Expected records: {TOTAL_PAYMENT_MODES:,}"
)

print(
    "✓ Payment mode records generated successfully."
)
```
