```python
"""
==============================================================================
Project : Hospital Analytics Portfolio Project
File    : 01_generate_departments.py
Purpose : Generate Department Dimension Data
Author  : Your Name
Version : 1.0
==============================================================================

Description
-----------
Generates the Hospital Department Master (Dimension) table.

Output
------
dim_department.csv

Total Records
-------------
18 Departments
==============================================================================
"""

# ==============================================================================
# IMPORT LIBRARIES
# ==============================================================================

import random
import pandas as pd

# ==============================================================================
# RANDOM SEED
# ==============================================================================

random.seed(100)

# ==============================================================================
# DEPARTMENT MASTER LIST
# ==============================================================================

DEPARTMENTS = [

    ("D001", "Cardiology",        "Heart related diseases"),
    ("D002", "Neurology",         "Brain and nervous system"),
    ("D003", "Orthopedics",       "Bones and joints"),
    ("D004", "Pediatrics",        "Child healthcare"),
    ("D005", "Gynecology",        "Women's healthcare"),
    ("D006", "Oncology",          "Cancer treatment"),
    ("D007", "Dermatology",       "Skin diseases"),
    ("D008", "ENT",               "Ear, Nose and Throat"),
    ("D009", "Ophthalmology",     "Eye care"),
    ("D010", "Psychiatry",        "Mental health"),
    ("D011", "Emergency",         "Emergency services"),
    ("D012", "Radiology",         "Medical imaging"),
    ("D013", "Pathology",         "Laboratory diagnosis"),
    ("D014", "Urology",           "Urinary system"),
    ("D015", "Nephrology",        "Kidney treatment"),
    ("D016", "Pulmonology",       "Respiratory diseases"),
    ("D017", "General Surgery",   "General surgical procedures"),
    ("D018", "Anesthesiology",    "Anesthesia services")

]

# ==============================================================================
# GENERATE DEPARTMENT DATA
# ==============================================================================

def generate_departments():
    """
    Generate department dimension records.

    Returns
    -------
    pandas.DataFrame
    """

    records = []

    for department_code, department_name, description in DEPARTMENTS:

        records.append({

            "department_code": department_code,

            "department_name": department_name,

            "department_description": description,

            "department_head":
                f"Dr. {random.choice(['A. Sharma','R. Patel','S. Mehta','P. Verma','K. Rao','N. Singh'])}",

            "floor_number":
                random.randint(1, 8),

            "phone_extension":
                random.randint(1000, 9999),

            "is_active":
                "Yes"

        })

    df = pd.DataFrame(records)

    return df


# ==============================================================================
# CREATE DATAFRAME
# ==============================================================================

department_df = generate_departments()

print("=" * 60)
print("Department Dimension Generated Successfully")
print("=" * 60)

print(f"Total Departments : {len(department_df)}")

print()

department_df.head()

# ==============================================================================
# VALIDATE DATA
# ==============================================================================

def validate_departments(df):
    """
    Validate Department Dimension Data.
    """

    print("\n" + "=" * 60)
    print("VALIDATING DEPARTMENT DATA")
    print("=" * 60)

    # Total Records
    print(f"Total Records           : {len(df)}")

    # Duplicate Department Codes
    duplicate_codes = df["department_code"].duplicated().sum()
    print(f"Duplicate Codes         : {duplicate_codes}")

    # Duplicate Department Names
    duplicate_names = df["department_name"].duplicated().sum()
    print(f"Duplicate Department    : {duplicate_names}")

    # Missing Values
    missing_values = df.isnull().sum().sum()
    print(f"Missing Values          : {missing_values}")

    # Active Departments
    active_departments = (df["is_active"] == "Yes").sum()
    print(f"Active Departments      : {active_departments}")

    print("=" * 60)

    if (
        duplicate_codes == 0
        and duplicate_names == 0
        and missing_values == 0
    ):
        print("Validation Status       : PASSED")
    else:
        print("Validation Status       : FAILED")

    print("=" * 60)


# ==============================================================================
# EXPORT TO CSV
# ==============================================================================

def export_to_csv(df):
    """
    Export Department Data to CSV.
    """

    file_name = "dim_department.csv"

    df.to_csv(
        file_name,
        index=False,
        encoding="utf-8"
    )

    print(f"\nCSV File Saved Successfully")
    print(f"Location : {file_name}")


# ==============================================================================
# OPTIONAL DATABASE INSERT
# ==============================================================================

def export_to_database(df):
    """
    Placeholder for MySQL Insert.

    This function will be connected with
    database_config.py in later parts.
    """

    print("\nDatabase Export")
    print("-" * 60)
    print("Skipped (Database integration will be added later.)")


# ==============================================================================
# SUMMARY REPORT
# ==============================================================================

def summary_report(df):
    """
    Display Summary Report.
    """

    print("\n" + "=" * 60)
    print("DEPARTMENT SUMMARY REPORT")
    print("=" * 60)

    print(df)

    print("\nDepartment Count")

    print("-" * 60)

    print(df["department_name"])

    print("=" * 60)


# ==============================================================================
# MAIN FUNCTION
# ==============================================================================

def main():

    print("\nGenerating Department Dimension...")

    department_df = generate_departments()

    validate_departments(department_df)

    export_to_csv(department_df)

    export_to_database(department_df)

    summary_report(department_df)

    print("\nDepartment Dimension Generation Completed Successfully.")


# ==============================================================================
# RUN SCRIPT
# ==============================================================================

if __name__ == "__main__":

    main()


# ==============================================================================
# END OF FILE
# ==============================================================================
```
