```python
# ==============================================================================
# IMPORT REQUIRED LIBRARIES
# ==============================================================================

import os
import random
import numpy as np
import pandas as pd

from faker import Faker
from datetime import datetime


# ==============================================================================
# INITIALIZE RANDOM GENERATORS
# ==============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

fake = Faker("en_IN")
Faker.seed(SEED)

print("Libraries imported successfully.")
print(f"Random seed: {SEED}")
```

```python
# ==============================================================================
# MEDICINE GENERATION CONFIGURATION
# ==============================================================================

TOTAL_MEDICINES = 300

OUTPUT_DIRECTORY = os.path.join(
    "03_Datasets",
    "CSV_Files"
)

CSV_FILE_PATH = os.path.join(
    OUTPUT_DIRECTORY,
    "dim_medicines.csv"
)

TODAY = datetime.now()


print("=" * 80)
print("MEDICINE GENERATION CONFIGURATION")
print("=" * 80)

print(
    f"Total medicines : {TOTAL_MEDICINES}"
)

print(
    f"Output path     : {CSV_FILE_PATH}"
)
```

```python
# ==============================================================================
# MEDICINE MASTER DATA
# ==============================================================================

MEDICINE_CATEGORIES = [
    "Analgesic",
    "Antibiotic",
    "Antacid",
    "Antihypertensive",
    "Antidiabetic",
    "Antihistamine",
    "Antiviral",
    "Antifungal",
    "Cardiac",
    "Respiratory",
    "Gastrointestinal",
    "Neurological",
    "Psychiatric",
    "Dermatological",
    "Vitamins and Supplements",
    "Hormonal",
    "Emergency Medicine",
    "Pediatric Medicine",
    "Oncology",
    "Ophthalmic"
]


DOSAGE_FORMS = [
    "Tablet",
    "Capsule",
    "Syrup",
    "Injection",
    "Cream",
    "Ointment",
    "Drops",
    "Inhaler",
    "Suspension",
    "Powder",
    "Gel",
    "Solution"
]


ROUTES_OF_ADMINISTRATION = [
    "Oral",
    "Intravenous",
    "Intramuscular",
    "Subcutaneous",
    "Topical",
    "Ophthalmic",
    "Inhalation",
    "Nasal"
]


MANUFACTURERS = [
    "Sun Pharma",
    "Cipla",
    "Dr. Reddy's Laboratories",
    "Lupin",
    "Aurobindo Pharma",
    "Mankind Pharma",
    "Torrent Pharmaceuticals",
    "Zydus Lifesciences",
    "Abbott India",
    "Alkem Laboratories",
    "Glenmark Pharmaceuticals",
    "Intas Pharmaceuticals",
    "Alembic Pharmaceuticals",
    "Biocon",
    "Ipca Laboratories"
]


STORAGE_CONDITIONS = [
    "Room Temperature",
    "Below 25°C",
    "Below 30°C",
    "Refrigerated",
    "Protect from Light",
    "Cool and Dry Place"
]


MEDICINE_STATUSES = [
    "Active",
    "Inactive",
    "Discontinued"
]


print("Medicine master data loaded successfully.")
```

```python
# ==============================================================================
# MEDICINE NAME MASTER DATA
# ==============================================================================

GENERIC_NAMES = [
    "Paracetamol",
    "Ibuprofen",
    "Amoxicillin",
    "Azithromycin",
    "Cefixime",
    "Ceftriaxone",
    "Metformin",
    "Glimepiride",
    "Amlodipine",
    "Losartan",
    "Telmisartan",
    "Atorvastatin",
    "Pantoprazole",
    "Omeprazole",
    "Rabeprazole",
    "Cetirizine",
    "Levocetirizine",
    "Montelukast",
    "Salbutamol",
    "Budesonide",
    "Ondansetron",
    "Domperidone",
    "Diclofenac",
    "Tramadol",
    "Aspirin",
    "Clopidogrel",
    "Furosemide",
    "Enalapril",
    "Metoprolol",
    "Dexamethasone",
    "Prednisolone",
    "Fluconazole",
    "Acyclovir",
    "Insulin",
    "Heparin",
    "Warfarin",
    "Gabapentin",
    "Pregabalin",
    "Sertraline",
    "Escitalopram"
]


STRENGTHS = [
    "5 mg",
    "10 mg",
    "20 mg",
    "25 mg",
    "40 mg",
    "50 mg",
    "75 mg",
    "100 mg",
    "150 mg",
    "200 mg",
    "250 mg",
    "500 mg",
    "625 mg",
    "650 mg",
    "750 mg",
    "1000 mg"
]


print(
    f"Generic medicines available : "
    f"{len(GENERIC_NAMES)}"
)

print(
    f"Strengths available         : "
    f"{len(STRENGTHS)}"
)
```

```python
# ==============================================================================
# MEDICINE CODE GENERATOR
# ==============================================================================

def generate_medicine_code(medicine_id):

    return f"MED{medicine_id:05d}"


print(
    generate_medicine_code(1)
)

print(
    generate_medicine_code(
        TOTAL_MEDICINES
    )
)
```

```python
# ==============================================================================
# GENERATE MEDICINE RECORDS
# ==============================================================================

medicine_records = []


for medicine_id in range(
    1,
    TOTAL_MEDICINES + 1
):

    # --------------------------------------------------------------------------
    # IDENTIFIERS
    # --------------------------------------------------------------------------

    medicine_code = generate_medicine_code(
        medicine_id
    )


    # --------------------------------------------------------------------------
    # MEDICINE NAMES
    # --------------------------------------------------------------------------

    generic_name = random.choice(
        GENERIC_NAMES
    )

    brand_name = (
        generic_name
        + " "
        + random.choice(
            [
                "Plus",
                "Care",
                "Med",
                "Life",
                "Max",
                "Heal",
                "Sure",
                "Advance"
            ]
        )
    )

    medicine_name = (
        brand_name
        + " "
        + str(
            random.choice(
                [
                    1,
                    2,
                    5,
                    10,
                    20,
                    25,
                    50,
                    100
                ]
            )
        )
    )


    # --------------------------------------------------------------------------
    # CATEGORY
    # --------------------------------------------------------------------------

    medicine_category = random.choice(
        MEDICINE_CATEGORIES
    )


    # --------------------------------------------------------------------------
    # DOSAGE FORM
    # --------------------------------------------------------------------------

    dosage_form = random.choice(
        DOSAGE_FORMS
    )


    # --------------------------------------------------------------------------
    # STRENGTH
    # --------------------------------------------------------------------------

    strength = random.choice(
        STRENGTHS
    )


    # --------------------------------------------------------------------------
    # ROUTE OF ADMINISTRATION
    # --------------------------------------------------------------------------

    route_of_administration = random.choice(
        ROUTES_OF_ADMINISTRATION
    )


    # --------------------------------------------------------------------------
    # MANUFACTURER
    # --------------------------------------------------------------------------

    manufacturer = random.choice(
        MANUFACTURERS
    )


    # --------------------------------------------------------------------------
    # UNIT PRICE
    # --------------------------------------------------------------------------

    unit_price = round(
        random.uniform(
            5,
            2500
        ),
        2
    )


    # --------------------------------------------------------------------------
    # PRESCRIPTION REQUIREMENT
    # --------------------------------------------------------------------------

    prescription_required = random.choices(
        [
            True,
            False
        ],
        weights=[
            65,
            35
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # CONTROLLED MEDICINE
    # --------------------------------------------------------------------------

    controlled_medicine = random.choices(
        [
            True,
            False
        ],
        weights=[
            5,
            95
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # STORAGE CONDITION
    # --------------------------------------------------------------------------

    if dosage_form == "Injection":

        storage_condition = random.choice(
            [
                "Refrigerated",
                "Below 25°C",
                "Protect from Light"
            ]
        )

    else:

        storage_condition = random.choice(
            STORAGE_CONDITIONS
        )


    # --------------------------------------------------------------------------
    # MEDICINE STATUS
    # --------------------------------------------------------------------------

    medicine_status = random.choices(
        MEDICINE_STATUSES,
        weights=[
            94,
            4,
            2
        ],
        k=1
    )[0]


    # --------------------------------------------------------------------------
    # AUDIT INFORMATION
    # --------------------------------------------------------------------------

    created_date = fake.date_time_between(
        start_date="-5y",
        end_date=TODAY
    )

    last_updated = fake.date_time_between(
        start_date=created_date,
        end_date=TODAY
    )


    # --------------------------------------------------------------------------
    # CREATE RECORD
    # --------------------------------------------------------------------------

    medicine_records.append(
        {
            "medicine_id": medicine_id,
            "medicine_code": medicine_code,
            "medicine_name": medicine_name,
            "generic_name": generic_name,
            "brand_name": brand_name,
            "medicine_category": medicine_category,
            "dosage_form": dosage_form,
            "strength": strength,
            "route_of_administration": route_of_administration,
            "manufacturer": manufacturer,
            "unit_price": unit_price,
            "prescription_required": prescription_required,
            "controlled_medicine": controlled_medicine,
            "storage_condition": storage_condition,
            "medicine_status": medicine_status,
            "created_date": created_date,
            "last_updated": last_updated
        }
    )


print(
    f"Generated {len(medicine_records):,} "
    f"medicine records."
)
```

```python
# ==============================================================================
# CREATE MEDICINE DATAFRAME
# ==============================================================================

medicines_df = pd.DataFrame(
    medicine_records
)


date_columns = [
    "created_date",
    "last_updated"
]


for column in date_columns:

    medicines_df[column] = pd.to_datetime(
        medicines_df[column]
    )


print("=" * 80)
print("MEDICINE DATAFRAME CREATED")
print("=" * 80)

print(
    f"Rows    : {medicines_df.shape[0]:,}"
)

print(
    f"Columns : {medicines_df.shape[1]}"
)

medicines_df.head()
```

```python
# ==============================================================================
# DATA TYPE CHECK
# ==============================================================================

print("=" * 80)
print("MEDICINE DATA TYPES")
print("=" * 80)

print(
    medicines_df.dtypes
)
```

```python
# ==============================================================================
# MEDICINE DATA VALIDATION
# ==============================================================================

print("=" * 80)
print("MEDICINE DATA VALIDATION")
print("=" * 80)


# ------------------------------------------------------------------------------
# 1. Row count
# ------------------------------------------------------------------------------

assert (
    len(medicines_df)
    == TOTAL_MEDICINES
)

print(
    "✓ Row count validation passed."
)


# ------------------------------------------------------------------------------
# 2. Medicine ID uniqueness
# ------------------------------------------------------------------------------

assert medicines_df[
    "medicine_id"
].is_unique

print(
    "✓ Medicine ID uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 3. Medicine code uniqueness
# ------------------------------------------------------------------------------

assert medicines_df[
    "medicine_code"
].is_unique

print(
    "✓ Medicine code uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 4. Medicine name uniqueness
# ------------------------------------------------------------------------------

assert medicines_df[
    "medicine_name"
].is_unique

print(
    "✓ Medicine name uniqueness validation passed."
)


# ------------------------------------------------------------------------------
# 5. Required fields
# ------------------------------------------------------------------------------

required_columns = [
    "medicine_id",
    "medicine_code",
    "medicine_name",
    "generic_name",
    "brand_name",
    "medicine_category",
    "dosage_form",
    "strength",
    "route_of_administration",
    "manufacturer",
    "unit_price",
    "medicine_status"
]


for column in required_columns:

    assert medicines_df[
        column
    ].notna().all(), (
        f"Null values found in {column}"
    )


print(
    "✓ Required field validation passed."
)


# ------------------------------------------------------------------------------
# 6. Category validation
# ------------------------------------------------------------------------------

assert medicines_df[
    "medicine_category"
].isin(
    MEDICINE_CATEGORIES
).all()

print(
    "✓ Medicine category validation passed."
)


# ------------------------------------------------------------------------------
# 7. Dosage form validation
# ------------------------------------------------------------------------------

assert medicines_df[
    "dosage_form"
].isin(
    DOSAGE_FORMS
).all()

print(
    "✓ Dosage form validation passed."
)


# ------------------------------------------------------------------------------
# 8. Route validation
# ------------------------------------------------------------------------------

assert medicines_df[
    "route_of_administration"
].isin(
    ROUTES_OF_ADMINISTRATION
).all()

print(
    "✓ Route validation passed."
)


# ------------------------------------------------------------------------------
# 9. Manufacturer validation
# ------------------------------------------------------------------------------

assert medicines_df[
    "manufacturer"
].isin(
    MANUFACTURERS
).all()

print(
    "✓ Manufacturer validation passed."
)


# ------------------------------------------------------------------------------
# 10. Financial validation
# ------------------------------------------------------------------------------

assert (
    medicines_df[
        "unit_price"
    ] > 0
).all()

print(
    "✓ Unit price validation passed."
)


# ------------------------------------------------------------------------------
# 11. Prescription flag validation
# ------------------------------------------------------------------------------

assert medicines_df[
    "prescription_required"
].isin(
    [
        True,
        False
    ]
).all()

print(
    "✓ Prescription flag validation passed."
)


# ------------------------------------------------------------------------------
# 12. Controlled medicine flag validation
# ------------------------------------------------------------------------------

assert medicines_df[
    "controlled_medicine"
].isin(
    [
        True,
        False
    ]
).all()

print(
    "✓ Controlled medicine validation passed."
)


# ------------------------------------------------------------------------------
# 13. Status validation
# ------------------------------------------------------------------------------

assert medicines_df[
    "medicine_status"
].isin(
    MEDICINE_STATUSES
).all()

print(
    "✓ Medicine status validation passed."
)


# ------------------------------------------------------------------------------
# 14. Medicine code format
# ------------------------------------------------------------------------------

assert medicines_df[
    "medicine_code"
].str.match(
    r"^MED\d{5}$"
).all()

print(
    "✓ Medicine code format validation passed."
)


print()

print(
    "All medicine data validations passed successfully. ✓"
)
```

```python
# ==============================================================================
# EXPORT MEDICINE DATA TO CSV
# ==============================================================================

os.makedirs(
    OUTPUT_DIRECTORY,
    exist_ok=True
)


def export_to_csv(medicines_df):
    """
    Export medicine dimension data into CSV.
    """

    medicines_df.to_csv(
        CSV_FILE_PATH,
        index=False
    )


    print("=" * 80)
    print("MEDICINE DATA EXPORT")
    print("=" * 80)


    print(
        "CSV file created successfully:"
    )


    print(
        CSV_FILE_PATH
    )


    return CSV_FILE_PATH


csv_file = export_to_csv(
    medicines_df
)
```

```python
# ==============================================================================
# VERIFY EXPORTED FILE
# ==============================================================================

assert os.path.exists(
    csv_file
)


file_size_kb = (
    os.path.getsize(csv_file)
    / 1024
)


print("=" * 80)
print("EXPORT VERIFICATION")
print("=" * 80)


print(
    f"File exists : "
    f"{os.path.exists(csv_file)}"
)


print(
    f"File path   : "
    f"{csv_file}"
)


print(
    f"File size   : "
    f"{file_size_kb:.2f} KB"
)
```

```python
# ==============================================================================
# MEDICINE CATEGORY PRICING SUMMARY
# ==============================================================================

print("=" * 80)
print("MEDICINE CATEGORY PRICING SUMMARY")
print("=" * 80)


category_price_summary = (
    medicines_df
    .groupby(
        "medicine_category"
    )
    .agg(
        medicine_count=(
            "medicine_id",
            "count"
        ),
        average_unit_price=(
            "unit_price",
            "mean"
        ),
        minimum_unit_price=(
            "unit_price",
            "min"
        ),
        maximum_unit_price=(
            "unit_price",
            "max"
        )
    )
    .reset_index()
)


category_price_summary[
    "average_unit_price"
] = category_price_summary[
    "average_unit_price"
].round(2)


category_price_summary
```

```python
# ==============================================================================
# PRESCRIPTION AND CONTROLLED MEDICINE SUMMARY
# ==============================================================================

print("=" * 80)
print("PRESCRIPTION AND CONTROLLED MEDICINE SUMMARY")
print("=" * 80)


prescription_summary = pd.DataFrame(
    {
        "Metric": [
            "Prescription Required",
            "Prescription Not Required",
            "Controlled Medicine",
            "Non-Controlled Medicine"
        ],

        "Count": [
            medicines_df[
                "prescription_required"
            ].sum(),

            (
                ~medicines_df[
                    "prescription_required"
                ]
            ).sum(),

            medicines_df[
                "controlled_medicine"
            ].sum(),

            (
                ~medicines_df[
                    "controlled_medicine"
                ]
            ).sum()
        ]
    }
)


prescription_summary
```
