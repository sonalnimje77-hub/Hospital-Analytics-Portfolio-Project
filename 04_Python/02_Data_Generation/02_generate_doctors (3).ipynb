```python
"""
==============================================================================
Project : Hospital Analytics Portfolio Project

File    : 02_generate_doctors.py

Purpose : Generate Doctor Dimension Data

Author  : Your Name

Version : 2.0

Description
-----------
This script generates realistic doctor master data for the
Hospital Analytics Portfolio Project.

Features
--------
✓ Generates 120 doctor records
✓ Indian doctor names
✓ Department mapping
✓ Qualification generation
✓ Specialization mapping
✓ Unique phone numbers
✓ Unique email addresses
✓ Medical license numbers
✓ Consultation fees
✓ Data validation
✓ CSV export
✓ Optional MySQL export
✓ Summary reporting

Output
------
dim_doctor.csv

Target Table
------------
dim_doctor

Compatible With
---------------
Python 3.10+
Jupyter Notebook
VS Code
MySQL 8
Power BI

==============================================================================
"""


# ==============================================================================
# INSTALL REQUIRED LIBRARIES
# ==============================================================================

import sys
import subprocess


def install_package(package):

    subprocess.check_call(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            package
        ]
    )


try:
    import pandas as pd

except ImportError:

    install_package("pandas")

    import pandas as pd



try:

    from faker import Faker

except ImportError:

    install_package("faker")

    from faker import Faker



try:

    import mysql.connector

except ImportError:

    install_package("mysql-connector-python")

    import mysql.connector



# ==============================================================================
# STANDARD LIBRARIES
# ==============================================================================

import random
import string
import os

from datetime import datetime, timedelta



# ==============================================================================
# RANDOM CONFIGURATION
# ==============================================================================

RANDOM_SEED = 100

random.seed(RANDOM_SEED)

Faker.seed(RANDOM_SEED)

fake = Faker("en_IN")



# ==============================================================================
# PROJECT CONFIGURATION
# ==============================================================================

PROJECT_NAME = "Hospital Analytics Portfolio Project"

DATASET_NAME = "Doctor Dimension"

TABLE_NAME = "dim_doctor"


TOTAL_DOCTORS = 120


CSV_FILE_NAME = "dim_doctor.csv"



# ==============================================================================
# DOCTOR CODE CONFIGURATION
# ==============================================================================

DOCTOR_CODE_PREFIX = "DOC"

DOCTOR_CODE_DIGITS = 4



# ==============================================================================
# AGE CONFIGURATION
# ==============================================================================

MIN_DOCTOR_AGE = 28

MAX_DOCTOR_AGE = 70



# ==============================================================================
# EXPERIENCE CONFIGURATION
# ==============================================================================

MIN_EXPERIENCE = 1

MAX_EXPERIENCE = 40



# ==============================================================================
# CONSULTATION FEE CONFIGURATION
# ==============================================================================

MIN_CONSULTATION_FEE = 300

MAX_CONSULTATION_FEE = 2500



# ==============================================================================
# MASTER LISTS
# ==============================================================================


GENDERS = [

    "Male",
    "Female"

]


SHIFT_TYPES = [

    "Morning",
    "Evening",
    "Night",
    "General"

]


EMPLOYMENT_TYPES = [

    "Full Time",
    "Part Time",
    "Consultant",
    "Visiting"

]


EMAIL_DOMAINS = [

    "gmail.com",
    "yahoo.com",
    "outlook.com",
    "hospital.org",
    "medicare.in"

]


MOBILE_PREFIX = [

    "98",
    "99",
    "97",
    "96",
    "95",
    "94",
    "93",
    "91"

]


LICENSE_PREFIX = "MCI"



# ==============================================================================
# DEPARTMENT MASTER
# ==============================================================================

DEPARTMENTS = [

    (1, "D001", "Cardiology"),

    (2, "D002", "Neurology"),

    (3, "D003", "Orthopedics"),

    (4, "D004", "Pediatrics"),

    (5, "D005", "Gynecology"),

    (6, "D006", "Oncology"),

    (7, "D007", "Dermatology"),

    (8, "D008", "ENT"),

    (9, "D009", "Ophthalmology"),

    (10, "D010", "Psychiatry"),

    (11, "D011", "Emergency"),

    (12, "D012", "Radiology"),

    (13, "D013", "Pathology"),

    (14, "D014", "Urology"),

    (15, "D015", "Nephrology"),

    (16, "D016", "Pulmonology"),

    (17, "D017", "General Surgery"),

    (18, "D018", "Anesthesiology")

]



DEPARTMENT_DF = pd.DataFrame(

    DEPARTMENTS,

    columns=[
        "department_id",
        "department_code",
        "department_name"
    ]

)



# ==============================================================================
# DEPARTMENT DOCTOR DISTRIBUTION
# ==============================================================================

DEPARTMENT_DOCTOR_COUNT = {


    "Cardiology":10,

    "Neurology":8,

    "Orthopedics":10,

    "Pediatrics":8,

    "Gynecology":9,

    "Oncology":7,

    "Dermatology":6,

    "ENT":6,

    "Ophthalmology":6,

    "Psychiatry":5,

    "Emergency":10,

    "Radiology":6,

    "Pathology":5,

    "Urology":5,

    "Nephrology":4,

    "Pulmonology":5,

    "General Surgery":7,

    "Anesthesiology":3

}



# Verification

assert sum(DEPARTMENT_DOCTOR_COUNT.values()) == TOTAL_DOCTORS



# ==============================================================================
# CONSULTATION FEE BY DEPARTMENT
# ==============================================================================


DEFAULT_CONSULTATION_FEES = {


    "Cardiology":1200,

    "Neurology":1500,

    "Orthopedics":1000,

    "Pediatrics":700,

    "Gynecology":900,

    "Oncology":1800,

    "Dermatology":600,

    "ENT":700,

    "Ophthalmology":800,

    "Psychiatry":1000,

    "Emergency":1000,

    "Radiology":900,

    "Pathology":600,

    "Urology":1200,

    "Nephrology":1300,

    "Pulmonology":1100,

    "General Surgery":1400,

    "Anesthesiology":1200

}



print("="*70)

print("Doctor Dimension Generator Initialized")

print("="*70)

print(f"Project       : {PROJECT_NAME}")

print(f"Target Table  : {TABLE_NAME}")

print(f"Doctors       : {TOTAL_DOCTORS}")

print("="*70)

# ==============================================================================
# QUALIFICATION MASTER
# ==============================================================================


QUALIFICATIONS = [

    "MBBS",

    "MBBS, MD",

    "MBBS, MS",

    "MBBS, DNB",

    "MBBS, MD, DM",

    "MBBS, MS, MCh",

    "MBBS, Fellowship"

]



# ==============================================================================
# DEPARTMENT-WISE SPECIALIZATION MASTER
# ==============================================================================


SPECIALIZATIONS = {


    "Cardiology":[

        "Interventional Cardiologist",

        "Clinical Cardiologist",

        "Electrophysiologist",

        "Heart Failure Specialist"

    ],


    "Neurology":[

        "Neurologist",

        "Stroke Specialist",

        "Epilepsy Specialist",

        "Neurophysician"

    ],


    "Orthopedics":[

        "Joint Replacement Surgeon",

        "Spine Specialist",

        "Trauma Surgeon",

        "Sports Medicine Specialist"

    ],


    "Pediatrics":[

        "General Pediatrician",

        "Neonatologist",

        "Pediatric Intensivist",

        "Pediatric Cardiologist"

    ],


    "Gynecology":[

        "Obstetrician",

        "Gynecologist",

        "Infertility Specialist",

        "Maternal Medicine Specialist"

    ],


    "Oncology":[

        "Medical Oncologist",

        "Surgical Oncologist",

        "Radiation Oncologist",

        "Cancer Specialist"

    ],


    "Dermatology":[

        "Dermatologist",

        "Cosmetic Dermatologist",

        "Skin Specialist",

        "Laser Specialist"

    ],


    "ENT":[

        "ENT Surgeon",

        "Otologist",

        "Rhinologist",

        "Laryngologist"

    ],


    "Ophthalmology":[

        "Eye Surgeon",

        "Retina Specialist",

        "Glaucoma Specialist",

        "Cornea Specialist"

    ],


    "Psychiatry":[

        "Psychiatrist",

        "Child Psychiatrist",

        "Clinical Psychiatrist",

        "Behavioral Specialist"

    ],


    "Emergency":[

        "Emergency Physician",

        "Trauma Specialist",

        "Critical Care Physician",

        "Emergency Consultant"

    ],


    "Radiology":[

        "Radiologist",

        "CT Scan Specialist",

        "MRI Specialist",

        "Ultrasound Specialist"

    ],


    "Pathology":[

        "Pathologist",

        "Clinical Pathologist",

        "Histopathologist",

        "Microbiologist"

    ],


    "Urology":[

        "Urologist",

        "Endourologist",

        "Kidney Stone Specialist",

        "Uro Surgeon"

    ],


    "Nephrology":[

        "Nephrologist",

        "Dialysis Specialist",

        "Kidney Specialist",

        "Renal Physician"

    ],


    "Pulmonology":[

        "Pulmonologist",

        "Chest Physician",

        "Respiratory Specialist",

        "Sleep Medicine Specialist"

    ],


    "General Surgery":[

        "General Surgeon",

        "Laparoscopic Surgeon",

        "GI Surgeon",

        "Trauma Surgeon"

    ],


    "Anesthesiology":[

        "Anesthesiologist",

        "Pain Management Specialist",

        "Critical Care Anesthetist",

        "Cardiac Anesthetist"

    ]

}



# ==============================================================================
# EXPERIENCE LEVEL MASTER
# ==============================================================================


EXPERIENCE_LEVELS = {


    "Junior":(1,5),

    "Mid-Level":(6,12),

    "Senior":(13,20),

    "Consultant":(21,30),

    "Chief Consultant":(31,40)

}



# ==============================================================================
# EXPERIENCE BASED FEE INCREMENT
# ==============================================================================


FEE_INCREMENT = {


    "Junior":0,

    "Mid-Level":200,

    "Senior":400,

    "Consultant":700,

    "Chief Consultant":1000

}



# ==============================================================================
# PREFERRED QUALIFICATION BY DEPARTMENT
# ==============================================================================


PREFERRED_QUALIFICATION = {


    "Cardiology":"MBBS, MD, DM",

    "Neurology":"MBBS, MD, DM",

    "Orthopedics":"MBBS, MS",

    "Pediatrics":"MBBS, MD",

    "Gynecology":"MBBS, MS",

    "Oncology":"MBBS, MD",

    "Dermatology":"MBBS, MD",

    "ENT":"MBBS, MS",

    "Ophthalmology":"MBBS, MS",

    "Psychiatry":"MBBS, MD",

    "Emergency":"MBBS, MD",

    "Radiology":"MBBS, MD",

    "Pathology":"MBBS, MD",

    "Urology":"MBBS, MS, MCh",

    "Nephrology":"MBBS, MD, DM",

    "Pulmonology":"MBBS, MD",

    "General Surgery":"MBBS, MS",

    "Anesthesiology":"MBBS, MD"

}



# ==============================================================================
# HELPER FUNCTIONS
# ==============================================================================


def generate_doctor_code(number):

    """
    Generate unique doctor code.

    Example:
    DOC0001
    """

    return f"DOC{number:04d}"



# ------------------------------------------------------------------------------

def generate_license_number():

    """
    Generate medical license number.
    """

    number = random.randint(
        100000,
        999999
    )

    return f"{LICENSE_PREFIX}{number}"



# ------------------------------------------------------------------------------

def generate_mobile_number():

    """
    Generate Indian mobile number.
    """

    prefix = random.choice(
        MOBILE_PREFIX
    )


    remaining = "".join(

        random.choices(
            string.digits,
            k=8
        )

    )


    return prefix + remaining



# ------------------------------------------------------------------------------

def generate_email(first_name,last_name):

    """
    Generate doctor email.
    """

    domain = random.choice(
        EMAIL_DOMAINS
    )


    return (

        f"{first_name.lower()}."
        f"{last_name.lower()}@{domain}"

    )



# ------------------------------------------------------------------------------

def generate_date_of_birth():

    """
    Generate DOB according to doctor age.
    """

    age = random.randint(

        MIN_DOCTOR_AGE,

        MAX_DOCTOR_AGE

    )


    dob = datetime.today() - timedelta(

        days=age*365

    )


    return dob.date()



# ------------------------------------------------------------------------------

def generate_experience():

    """
    Generate years of experience.
    """

    return random.randint(

        MIN_EXPERIENCE,

        MAX_EXPERIENCE

    )



# ------------------------------------------------------------------------------

def get_experience_level(years):

    """
    Convert years into experience category.
    """

    for level,limit in EXPERIENCE_LEVELS.items():


        if limit[0] <= years <= limit[1]:

            return level


    return "Junior"



# ------------------------------------------------------------------------------

def generate_joining_date(experience):

    """
    Generate approximate joining date.
    """

    year = datetime.now().year - experience


    month = random.randint(
        1,
        12
    )


    day = random.randint(
        1,
        28
    )


    return datetime(

        year,
        month,
        day

    ).date()



# ------------------------------------------------------------------------------

def generate_consultation_fee(
        department,
        experience_level
):

    """
    Generate consultation fee.
    """

    base_fee = DEFAULT_CONSULTATION_FEES.get(

        department,

        MIN_CONSULTATION_FEE

    )


    increment = FEE_INCREMENT.get(

        experience_level,

        0

    )


    return base_fee + increment



# ------------------------------------------------------------------------------

def generate_specialization(department):

    """
    Generate department specific specialization.
    """

    return random.choice(

        SPECIALIZATIONS[department]

    )



# ------------------------------------------------------------------------------

def generate_qualification(department):

    """
    Generate qualification.
    """

    return PREFERRED_QUALIFICATION.get(

        department,

        random.choice(QUALIFICATIONS)

    )



# ------------------------------------------------------------------------------

def generate_shift():

    return random.choice(

        SHIFT_TYPES

    )



# ------------------------------------------------------------------------------

def generate_employment_type():

    return random.choice(

        EMPLOYMENT_TYPES

    )



# ------------------------------------------------------------------------------

def generate_status():

    return random.choices(

        ["Active","Inactive"],

        weights=[96,4],

        k=1

    )[0]



print("\nHelper Functions Loaded Successfully")

print("-"*70)

print("Doctor Code Generator       : OK")

print("License Generator           : OK")

print("Mobile Generator            : OK")

print("Email Generator             : OK")

print("DOB Generator               : OK")

print("Experience Generator        : OK")

print("Fee Generator               : OK")

print("Specialization Generator    : OK")

print("-"*70)

# ==============================================================================
# GENERATE DOCTOR DIMENSION DATA
# ==============================================================================


def generate_doctors():

    """
    Generate Doctor Dimension Data.

    Returns
    -------
    pandas.DataFrame
        Generated doctor master data
    """


    print("\n" + "="*70)

    print("GENERATING DOCTOR DIMENSION DATA")

    print("="*70)



    doctor_records = []



    used_doctor_codes = set()

    used_license_numbers = set()

    used_mobile_numbers = set()

    used_email_addresses = set()



    doctor_counter = 1



    # ==========================================================================
    # DEPARTMENT-WISE GENERATION
    # ==========================================================================


    for department_name, doctor_count in DEPARTMENT_DOCTOR_COUNT.items():


        print(
            f"Generating {doctor_count} doctors for {department_name}"
        )



        department_row = DEPARTMENT_DF[

            DEPARTMENT_DF["department_name"]

            == department_name

        ].iloc[0]



        department_id = int(

            department_row["department_id"]

        )


        department_code = department_row[

            "department_code"

        ]



        # ----------------------------------------------------------------------
        # CREATE DOCTORS
        # ----------------------------------------------------------------------


        for _ in range(doctor_count):



            # ------------------------------------------------------------------
            # Doctor Code
            # ------------------------------------------------------------------


            doctor_code = generate_doctor_code(

                doctor_counter

            )


            while doctor_code in used_doctor_codes:


                doctor_counter += 1


                doctor_code = generate_doctor_code(

                    doctor_counter

                )


            used_doctor_codes.add(

                doctor_code

            )



            # ------------------------------------------------------------------
            # Name Generation
            # ------------------------------------------------------------------


            gender = random.choice(

                GENDERS

            )



            if gender == "Male":


                first_name = fake.first_name_male()


            else:


                first_name = fake.first_name_female()



            last_name = fake.last_name()



            # ------------------------------------------------------------------
            # Professional Information
            # ------------------------------------------------------------------


            date_of_birth = generate_date_of_birth()



            experience_years = generate_experience()



            experience_level = get_experience_level(

                experience_years

            )



            qualification = generate_qualification(

                department_name

            )



            specialization = generate_specialization(

                department_name

            )



            consultation_fee = generate_consultation_fee(

                department_name,

                experience_level

            )



            # ------------------------------------------------------------------
            # Mobile Number
            # ------------------------------------------------------------------


            phone_number = generate_mobile_number()



            while phone_number in used_mobile_numbers:


                phone_number = generate_mobile_number()



            used_mobile_numbers.add(

                phone_number

            )



            # ------------------------------------------------------------------
            # Email
            # ------------------------------------------------------------------


            email = generate_email(

                first_name,

                last_name

            )



            counter = 1



            while email in used_email_addresses:


                email = generate_email(

                    first_name,

                    f"{last_name}{counter}"

                )


                counter += 1



            used_email_addresses.add(

                email

            )



            # ------------------------------------------------------------------
            # Medical License
            # ------------------------------------------------------------------


            license_number = generate_license_number()



            while license_number in used_license_numbers:


                license_number = generate_license_number()



            used_license_numbers.add(

                license_number

            )



            # ------------------------------------------------------------------
            # Employment Details
            # ------------------------------------------------------------------


            joining_date = generate_joining_date(

                experience_years

            )


            shift_type = generate_shift()



            employment_type = generate_employment_type()



            doctor_status = generate_status()



            # ------------------------------------------------------------------
            # Create Record
            # ------------------------------------------------------------------


            doctor_record = {


                "doctor_id":

                    None,


                "doctor_code":

                    doctor_code,


                "first_name":

                    first_name,


                "last_name":

                    last_name,


                "gender":

                    gender,


                "date_of_birth":

                    date_of_birth,


                "specialization":

                    specialization,


                "qualification":

                    qualification,


                "experience_years":

                    experience_years,


                "experience_level":

                    experience_level,


                "department_id":

                    department_id,


                "department_code":

                    department_code,


                "department_name":

                    department_name,


                "consultation_fee":

                    consultation_fee,


                "phone_number":

                    phone_number,


                "email":

                    email,


                "joining_date":

                    joining_date,


                "shift_type":

                    shift_type,


                "employment_type":

                    employment_type,


                "doctor_status":

                    doctor_status,


                "license_number":

                    license_number,


                "created_date":

                    datetime.now().date(),


                "last_updated":

                    datetime.now()

            }



            doctor_records.append(

                doctor_record

            )


            doctor_counter += 1



    # ==========================================================================
    # CREATE DATAFRAME
    # ==========================================================================


    doctor_df = pd.DataFrame(

        doctor_records

    )



    # ==========================================================================
    # COLUMN ORDER
    # ==========================================================================


    column_order = [


        "doctor_id",

        "doctor_code",

        "first_name",

        "last_name",

        "gender",

        "date_of_birth",

        "specialization",

        "qualification",

        "experience_years",

        "experience_level",

        "department_id",

        "department_code",

        "department_name",

        "consultation_fee",

        "phone_number",

        "email",

        "joining_date",

        "shift_type",

        "employment_type",

        "doctor_status",

        "license_number",

        "created_date",

        "last_updated"

    ]



    doctor_df = doctor_df[

        column_order

    ]



    doctor_df.sort_values(

        by="doctor_code",

        inplace=True

    )



    doctor_df.reset_index(

        drop=True,

        inplace=True

    )



    print("\nDoctor Data Generated Successfully")

    print("-"*70)

    print(

        f"Total Doctors Generated : {len(doctor_df)}"

    )

    print(

        f"Total Columns           : {len(doctor_df.columns)}"

    )

    print("-"*70)



    print("\nSample Data")

    print("-"*70)

    print(

        doctor_df.head()

    )



    return doctor_df





# ==============================================================================
# VALIDATE DOCTOR DATA
# ==============================================================================


def validate_doctor_data(doctor_df):

    """
    Perform data quality validation.
    """


    print("\n" + "="*70)

    print("VALIDATING DOCTOR DIMENSION DATA")

    print("="*70)



    validation_passed = True



    # --------------------------------------------------------------------------
    # Record Count
    # --------------------------------------------------------------------------


    print(

        f"Total Records : {len(doctor_df)}"

    )



    if len(doctor_df) != TOTAL_DOCTORS:


        print("Record Count Validation FAILED")

        validation_passed = False


    else:


        print("Record Count Validation PASSED")



    print("-"*70)



    # --------------------------------------------------------------------------
    # Duplicate Checks
    # --------------------------------------------------------------------------


    duplicate_checks = {


        "Doctor Codes":

            doctor_df["doctor_code"].duplicated().sum(),


        "License Numbers":

            doctor_df["license_number"].duplicated().sum(),


        "Phone Numbers":

            doctor_df["phone_number"].duplicated().sum(),


        "Email Addresses":

            doctor_df["email"].duplicated().sum()

    }



    for key,value in duplicate_checks.items():


        print(

            f"{key:<25}: {value}"

        )


        if value > 0:

            validation_passed = False



    print("-"*70)



    # --------------------------------------------------------------------------
    # Missing Values
    # --------------------------------------------------------------------------


    missing_values = doctor_df.isnull().sum().sum()



    print(

        f"Missing Values : {missing_values}"

    )



    if missing_values > 0:


        validation_passed = False



    print("-"*70)



    # --------------------------------------------------------------------------
    # Department Summary
    # --------------------------------------------------------------------------


    print("\nDepartment Wise Doctor Count")

    print("-"*70)



    print(

        doctor_df

        .groupby("department_name")

        .size()

    )



    print("-"*70)



    if validation_passed:


        print(

            "VALIDATION RESULT : PASSED"

        )


    else:


        print(

            "VALIDATION RESULT : FAILED"

        )



    print("="*70)



    return validation_passed





# ==============================================================================
# CREATE DOCTOR DATAFRAME
# ==============================================================================


doctor_df = generate_doctors()



validation_result = validate_doctor_data(

    doctor_df

)

# ==============================================================================
# EXPORT DOCTOR DATA TO CSV
# ==============================================================================


def export_to_csv(doctor_df):

    """
    Export doctor dimension data into CSV file.
    """


    print("\n" + "="*70)

    print("EXPORTING DOCTOR DATA TO CSV")

    print("="*70)



    output_folder = "03_Datasets/CSV_Files"



    os.makedirs(

        output_folder,

        exist_ok=True

    )



    csv_path = os.path.join(

        output_folder,

        "dim_doctor.csv"

    )



    try:


        doctor_df.to_csv(

            csv_path,

            index=False,

            encoding="utf-8"

        )



        file_size = os.path.getsize(

            csv_path

        ) / 1024



        print(

            "CSV Export Status : SUCCESS"

        )


        print(

            f"File Location     : {csv_path}"

        )


        print(

            f"Records Exported  : {len(doctor_df)}"

        )


        print(

            f"Columns           : {len(doctor_df.columns)}"

        )


        print(

            f"File Size         : {file_size:.2f} KB"

        )



    except Exception as error:


        print(

            "CSV Export Failed"

        )


        print(error)


        return None



    print("="*70)



    return csv_path





# ==============================================================================
# VERIFY CSV FILE
# ==============================================================================


def verify_csv_export(csv_path):

    """
    Verify exported CSV file.
    """


    print("\nVerifying CSV File")

    print("-"*70)



    if csv_path is None:


        print(

            "CSV Verification Failed"

        )

        return False



    if not os.path.exists(csv_path):


        print(

            "CSV File Not Found"

        )

        return False



    try:


        df_check = pd.read_csv(

            csv_path

        )



        print(

            "CSV Verification SUCCESS"

        )


        print(

            f"Rows    : {len(df_check)}"

        )


        print(

            f"Columns : {len(df_check.columns)}"

        )



        return True



    except Exception as error:


        print(error)

        return False





# ==============================================================================
# MYSQL DATABASE CONFIGURATION
# ==============================================================================


DB_CONFIG = {


    "host":

        "localhost",


    "port":

        3306,


    "database":

        "hospital_analytics_db",


    "user":

        "root",


    "password":

        "YOUR_PASSWORD"


}





# ==============================================================================
# EXPORT DATA TO MYSQL
# ==============================================================================


def export_to_mysql(doctor_df):

    """
    Insert doctor dimension data into MySQL.
    """


    print("\n" + "="*70)

    print("EXPORTING DATA TO MYSQL")

    print("="*70)



    connection = None

    cursor = None



    try:


        connection = mysql.connector.connect(

            **DB_CONFIG

        )



        cursor = connection.cursor()



        print(

            "Database Connected Successfully"

        )



        insert_query = """


        INSERT INTO dim_doctor

        (

            doctor_code,

            first_name,

            last_name,

            gender,

            date_of_birth,

            specialization,

            qualification,

            experience_years,

            experience_level,

            department_id,

            consultation_fee,

            phone_number,

            email,

            joining_date,

            shift_type,

            employment_type,

            doctor_status,

            license_number

        )


        VALUES

        (

            %s,%s,%s,%s,%s,%s,%s,%s,

            %s,%s,%s,%s,%s,%s,%s,%s,

            %s,%s

        )

        """



        inserted = 0



        for _,row in doctor_df.iterrows():



            values = (


                row["doctor_code"],


                row["first_name"],


                row["last_name"],


                row["gender"],


                row["date_of_birth"],


                row["specialization"],


                row["qualification"],


                int(row["experience_years"]),


                row["experience_level"],


                int(row["department_id"]),


                float(row["consultation_fee"]),


                row["phone_number"],


                row["email"],


                row["joining_date"],


                row["shift_type"],


                row["employment_type"],


                row["doctor_status"],


                row["license_number"]

            )



            cursor.execute(

                insert_query,

                values

            )



            inserted += 1




        connection.commit()



        print("-"*70)

        print(

            "MYSQL EXPORT SUCCESS"

        )


        print(

            f"Inserted Records : {inserted}"

        )


        print("-"*70)



    except Exception as error:


        print(

            "MYSQL EXPORT FAILED"

        )


        print(error)



    finally:



        if cursor:


            cursor.close()



        if connection and connection.is_connected():


            connection.close()


            print(

                "Database Connection Closed"

            )





# ==============================================================================
# VERIFY MYSQL EXPORT
# ==============================================================================


def verify_mysql_export():

    """
    Verify records inserted into MySQL.
    """


    print("\n" + "="*70)

    print("VERIFYING MYSQL DATA")

    print("="*70)



    connection = None

    cursor = None



    try:


        connection = mysql.connector.connect(

            **DB_CONFIG

        )


        cursor = connection.cursor()



        cursor.execute(

            """
            SELECT COUNT(*)
            FROM dim_doctor
            """

        )



        total = cursor.fetchone()[0]



        print(

            f"Total Records in dim_doctor : {total}"

        )



        print("\nDepartment Wise Doctor Count")

        print("-"*50)



        cursor.execute(

            """

            SELECT

                department_id,

                COUNT(*)

            FROM dim_doctor

            GROUP BY department_id

            ORDER BY department_id

            """

        )



        for row in cursor.fetchall():


            print(

                f"Department {row[0]} : {row[1]} Doctors"

            )



    except Exception as error:


        print(error)



    finally:



        if cursor:


            cursor.close()



        if connection and connection.is_connected():


            connection.close()



# ==============================================================================
# EXECUTE EXPORT PROCESS
# ==============================================================================


csv_file = export_to_csv(

    doctor_df

)



verify_csv_export(

    csv_file

)



# Uncomment below after MySQL setup

# export_to_mysql(doctor_df)

# verify_mysql_export()

# ==============================================================================
# GENERATE SUMMARY REPORT
# ==============================================================================


def generate_summary_report(doctor_df):

    """
    Generate professional execution summary.
    """


    print("\n")

    print("="*80)

    print("        DOCTOR DIMENSION SUMMARY REPORT")

    print("="*80)



    # --------------------------------------------------------------------------
    # GENERAL INFORMATION
    # --------------------------------------------------------------------------


    print("\nGENERAL INFORMATION")

    print("-"*80)



    print(

        f"Project Name          : {PROJECT_NAME}"

    )


    print(

        f"Dataset               : {DATASET_NAME}"

    )


    print(

        f"Target Table          : {TABLE_NAME}"

    )


    print(

        f"Total Doctors         : {len(doctor_df):,}"

    )


    print(

        f"Total Columns         : {len(doctor_df.columns)}"

    )


    print(

        f"Departments Covered   : {doctor_df['department_name'].nunique()}"

    )




    # --------------------------------------------------------------------------
    # GENDER DISTRIBUTION
    # --------------------------------------------------------------------------


    print("\nGENDER DISTRIBUTION")

    print("-"*80)



    gender_summary = (

        doctor_df["gender"]

        .value_counts()

    )



    for gender,count in gender_summary.items():


        percentage = (

            count / len(doctor_df)

        ) * 100



        print(

            f"{gender:<15} {count:>5} ({percentage:.2f}%)"

        )




    # --------------------------------------------------------------------------
    # DEPARTMENT DISTRIBUTION
    # --------------------------------------------------------------------------


    print("\nDEPARTMENT WISE DOCTOR COUNT")

    print("-"*80)



    department_summary = (

        doctor_df

        .groupby("department_name")

        .size()

        .sort_values(

            ascending=False

        )

    )



    for department,total in department_summary.items():


        print(

            f"{department:<30} {total:>5}"

        )




    # --------------------------------------------------------------------------
    # EXPERIENCE STATISTICS
    # --------------------------------------------------------------------------


    print("\nEXPERIENCE STATISTICS")

    print("-"*80)



    print(

        f"Minimum Experience : {doctor_df['experience_years'].min()} Years"

    )


    print(

        f"Maximum Experience : {doctor_df['experience_years'].max()} Years"

    )


    print(

        f"Average Experience : {doctor_df['experience_years'].mean():.2f} Years"

    )




    # --------------------------------------------------------------------------
    # CONSULTATION FEE STATISTICS
    # --------------------------------------------------------------------------


    print("\nCONSULTATION FEE STATISTICS")

    print("-"*80)



    print(

        f"Minimum Fee : ₹ {doctor_df['consultation_fee'].min():,.2f}"

    )


    print(

        f"Maximum Fee : ₹ {doctor_df['consultation_fee'].max():,.2f}"

    )


    print(

        f"Average Fee : ₹ {doctor_df['consultation_fee'].mean():,.2f}"

    )




    # --------------------------------------------------------------------------
    # DATA QUALITY SUMMARY
    # --------------------------------------------------------------------------


    print("\nDATA QUALITY CHECK")

    print("-"*80)



    print(

        "Duplicate Doctor Codes :",

        doctor_df["doctor_code"].duplicated().sum()

    )


    print(

        "Duplicate Emails       :",

        doctor_df["email"].duplicated().sum()

    )


    print(

        "Duplicate Phones       :",

        doctor_df["phone_number"].duplicated().sum()

    )


    print(

        "Duplicate Licenses     :",

        doctor_df["license_number"].duplicated().sum()

    )


    print(

        "Missing Values         :",

        doctor_df.isnull().sum().sum()

    )




    # --------------------------------------------------------------------------
    # SAMPLE DATA
    # --------------------------------------------------------------------------


    print("\nFIRST FIVE RECORDS")

    print("-"*80)


    print(

        doctor_df.head()

    )



    print("\n")

    print("="*80)

    print("SUMMARY REPORT GENERATED SUCCESSFULLY")

    print("="*80)





# ==============================================================================
# MAIN PIPELINE FUNCTION
# ==============================================================================


def main():

    """
    Execute complete doctor dimension pipeline.
    """



    start_time = datetime.now()



    print("\n")

    print("="*80)

    print("HOSPITAL ANALYTICS PORTFOLIO PROJECT")

    print("DOCTOR DIMENSION DATA PIPELINE")

    print("="*80)



    print(

        f"Started At : {start_time}"

    )




    # --------------------------------------------------------------------------
    # STEP 1 : GENERATE DATA
    # --------------------------------------------------------------------------


    print("\nSTEP 1/5 : GENERATING DOCTOR DATA")

    print("-"*80)



    doctor_df = generate_doctors()




    # --------------------------------------------------------------------------
    # STEP 2 : VALIDATION
    # --------------------------------------------------------------------------


    print("\nSTEP 2/5 : VALIDATING DATA")

    print("-"*80)



    validation_result = validate_doctor_data(

        doctor_df

    )



    if validation_result is False:


        print(

            "Validation Failed. Pipeline Stopped."

        )


        return None




    # --------------------------------------------------------------------------
    # STEP 3 : CSV EXPORT
    # --------------------------------------------------------------------------


    print("\nSTEP 3/5 : EXPORTING CSV")

    print("-"*80)



    csv_file = export_to_csv(

        doctor_df

    )



    verify_csv_export(

        csv_file

    )




    # --------------------------------------------------------------------------
    # STEP 4 : MYSQL EXPORT
    # --------------------------------------------------------------------------


    print("\nSTEP 4/5 : MYSQL EXPORT")

    print("-"*80)



    print(

        "MySQL Export Disabled By Default"

    )


    print(

        "Enable manually after database configuration"

    )



    # Uncomment after MySQL setup

    # export_to_mysql(doctor_df)

    # verify_mysql_export()




    # --------------------------------------------------------------------------
    # STEP 5 : SUMMARY
    # --------------------------------------------------------------------------


    print("\nSTEP 5/5 : SUMMARY REPORT")

    print("-"*80)



    generate_summary_report(

        doctor_df

    )




    end_time = datetime.now()



    print("\n")

    print("="*80)

    print("PIPELINE COMPLETED SUCCESSFULLY")

    print("="*80)



    print(

        f"Start Time : {start_time}"

    )


    print(

        f"End Time   : {end_time}"

    )


    print(

        f"Duration   : {end_time-start_time}"

    )


    print(

        f"Doctors Generated : {len(doctor_df)}"

    )


    print("="*80)



    return doctor_df





# ==============================================================================
# SCRIPT EXECUTION
# ==============================================================================


if __name__ == "__main__":


    try:


        doctor_df = main()



        if doctor_df is not None:


            print("\n")

            print("="*80)

            print("DOCTOR DIMENSION READY")

            print("="*80)


            print(

                "✓ CSV Dataset Created"

            )


            print(

                "✓ Data Validation Passed"

            )


            print(

                "✓ Power BI Ready"

            )


            print(

                "✓ Excel Ready"

            )


            print(

                "✓ SQL Import Ready"

            )


            print("="*80)



    except KeyboardInterrupt:


        print("\nProcess Cancelled By User")



    except Exception as error:


        print("\nUnexpected Error Occurred")

        print("-"*80)

        print(error)



    finally:


        print("\n")

        print("="*80)

        print("END OF 02_generate_doctors.py")

        print("="*80)
```
