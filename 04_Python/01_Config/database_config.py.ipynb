```python
"""
==============================================================================
Project : Hospital Analytics Portfolio Project
File    : database_config.py
Purpose : MySQL Database Configuration
Author  : Your Name
Version : 1.0
==============================================================================

This file provides reusable MySQL database connection functions.

Functions Included
------------------
1. get_connection()
2. test_connection()

Used By
--------
• Data Generation Scripts
• Data Cleaning Scripts
• EDA Scripts
• Export Scripts
• Validation Scripts
==============================================================================
"""

# ==============================================================================
# IMPORT LIBRARIES
# ==============================================================================

try:
    import mysql.connector
    from mysql.connector import Error

except ImportError:

    import sys
    import subprocess

    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "mysql-connector-python"]
    )

    import mysql.connector
    from mysql.connector import Error

# ==============================================================================
# DATABASE CONFIGURATION
# ==============================================================================

DB_CONFIG = {

    "host": "localhost",

    "port": 3306,

    "database": "hospital_analytics_db",

    "user": "root",

    # Replace with your MySQL password
    "password": "YOUR_PASSWORD",

    "charset": "utf8mb4",

    "autocommit": True

}

# ==============================================================================
# CREATE DATABASE CONNECTION
# ==============================================================================

def get_connection():
    """
    Creates and returns a MySQL database connection.
    """

    try:

        connection = mysql.connector.connect(

            host=DB_CONFIG["host"],

            port=DB_CONFIG["port"],

            database=DB_CONFIG["database"],

            user=DB_CONFIG["user"],

            password=DB_CONFIG["password"],

            charset=DB_CONFIG["charset"],

            autocommit=DB_CONFIG["autocommit"]

        )

        return connection

    except Error as error:

        print("\n" + "=" * 60)
        print("DATABASE CONNECTION FAILED")
        print("=" * 60)

        if error.errno == 1045:

            print("Reason : Invalid username or password.")
            print()
            print("Please verify:")
            print("1. MySQL username")
            print("2. MySQL password")
            print("3. MySQL Server is running")
            print("4. Use the same credentials as MySQL Workbench")

        elif error.errno == 1049:

            print("Reason : Database does not exist.")
            print()
            print("Run the following SQL first:")
            print("CREATE DATABASE hospital_analytics_db;")

        elif error.errno == 2003:

            print("Reason : MySQL Server is not running.")
            print()
            print("Start MySQL Server and try again.")

        else:

            print(error)

        print("=" * 60)

        return None


# ==============================================================================
# TEST DATABASE CONNECTION
# ==============================================================================

def test_connection():
    """
    Tests the MySQL connection.
    Displays server information.
    """

    connection = get_connection()

    if connection is None:
        return

    try:

        cursor = connection.cursor()

        cursor.execute("SELECT DATABASE();")

        database_name = cursor.fetchone()[0]

        cursor.execute("SELECT VERSION();")

        mysql_version = cursor.fetchone()[0]

        print("=" * 60)
        print("DATABASE CONNECTION SUCCESSFUL")
        print("=" * 60)
        print(f"Database : {database_name}")
        print(f"MySQL    : {mysql_version}")
        print(f"Server   : {DB_CONFIG['host']}")
        print(f"Port     : {DB_CONFIG['port']}")
        print("=" * 60)

        cursor.close()

    except Error as error:

        print(f"Error while testing connection : {error}")

    finally:

        if connection is not None:

            if connection.is_connected():

                connection.close()

                print("Database connection closed.")

# ==============================================================================
# CLOSE DATABASE CONNECTION
# ==============================================================================

def close_connection(connection):
    """
    Safely closes the MySQL database connection.
    """

    if connection is not None:

        if connection.is_connected():

            connection.close()

            print("Database connection closed.")


# ==============================================================================
# EXECUTE SELECT QUERY
# ==============================================================================

def execute_select(query, values=None):
    """
    Executes a SELECT query and returns all rows.

    Parameters
    ----------
    query : str
        SQL SELECT statement

    values : tuple, optional
        Query parameters

    Returns
    -------
    list
        List of dictionaries containing query results
    """

    connection = get_connection()

    if connection is None:
        return []

    try:

        cursor = connection.cursor(dictionary=True)

        if values is None:

            cursor.execute(query)

        else:

            cursor.execute(query, values)

        results = cursor.fetchall()

        cursor.close()

        return results

    except Error as error:

        print(f"Query Error : {error}")

        return []

    finally:

        close_connection(connection)


# ==============================================================================
# EXECUTE INSERT / UPDATE / DELETE / DDL
# ==============================================================================

def execute_query(query, values=None):
    """
    Executes INSERT, UPDATE, DELETE,
    CREATE, ALTER and DROP statements.
    """

    connection = get_connection()

    if connection is None:
        return False

    try:

        cursor = connection.cursor()

        if values is None:

            cursor.execute(query)

        else:

            cursor.execute(query, values)

        connection.commit()

        print("Query executed successfully.")

        cursor.close()

        return True

    except Error as error:

        print(f"Execution Error : {error}")

        return False

    finally:

        close_connection(connection)


# ==============================================================================
# EXECUTE BULK INSERT
# ==============================================================================

def execute_many(query, data):
    """
    Executes bulk INSERT operations.

    Parameters
    ----------
    query : str
        INSERT statement

    data : list
        List of tuples
    """

    connection = get_connection()

    if connection is None:
        return False

    try:

        cursor = connection.cursor()

        cursor.executemany(query, data)

        connection.commit()

        print(f"{cursor.rowcount} rows inserted successfully.")

        cursor.close()

        return True

    except Error as error:

        print(f"Bulk Insert Error : {error}")

        return False

    finally:

        close_connection(connection)


# ==============================================================================
# SHOW DATABASE CONFIGURATION
# ==============================================================================

def show_database_config():
    """
    Displays the current database configuration.
    """

    print("\n" + "=" * 60)
    print("DATABASE CONFIGURATION")
    print("=" * 60)

    print(f"Host      : {DB_CONFIG['host']}")
    print(f"Port      : {DB_CONFIG['port']}")
    print(f"Database  : {DB_CONFIG['database']}")
    print(f"User      : {DB_CONFIG['user']}")
    print(f"Charset   : {DB_CONFIG['charset']}")

    print("=" * 60)


# ==============================================================================
# MAIN
# ==============================================================================

if __name__ == "__main__":

    show_database_config()

    print()

    test_connection()


# ==============================================================================
# END OF FILE
# ==============================================================================
```
